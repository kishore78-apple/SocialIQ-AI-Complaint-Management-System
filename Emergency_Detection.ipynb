```python
import os
import re
import getpass
import joblib
import warnings
import pandas as pd
import matplotlib.pyplot as plt

from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")


mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"
database_name = "government_analytics"
source_table = "emergency_detection"


mysql_password = getpass.getpass(
    "Enter MySQL password: "
)

encoded_password = quote_plus(mysql_password)


server_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)


with server_engine.begin() as connection:
    connection.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS "
            f"`{database_name}`"
        )
    )


database_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/"
    f"{database_name}",
    pool_pre_ping=True
)


with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()


print("Connected database:", connected_database)


existing_tables = inspect(
    database_engine
).get_table_names()


if source_table in existing_tables:

    emergency_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print("Dataset loaded from MySQL.")

else:

    possible_paths = [
        "emergency_detection.csv",
        "emergency_detection(1).csv",
        "emergency_detection(2).csv",
        "Emergency_Detection.csv",
        "/mnt/data/emergency_detection.csv",
        "/mnt/data/emergency_detection(1).csv",
        "/mnt/data/emergency_detection(2).csv",
        r"C:\Users\Kishore\Downloads\emergency_detection.csv",
        r"C:\Users\Kishore\Downloads\emergency_detection(1).csv",
        r"C:\Users\Kishore\Downloads\emergency_detection(2).csv"
    ]

    csv_path = next(
        (
            path
            for path in possible_paths
            if os.path.exists(path)
        ),
        None
    )

    if csv_path is None:

        raise FileNotFoundError(
            "Place emergency_detection.csv "
            "in the same folder as the notebook."
        )

    emergency_df = pd.read_csv(
        csv_path,
        low_memory=False
    )

    emergency_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print(
        "Dataset loaded from CSV and uploaded to MySQL."
    )


if emergency_df.empty:
    raise ValueError("The dataset is empty.")


print("\nOriginal dataset shape:", emergency_df.shape)

print("\nOriginal columns:")
print(emergency_df.columns.tolist())

display(emergency_df.head())


emergency_df.columns = (
    emergency_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(
        r"[^a-z0-9]+",
        "_",
        regex=True
    )
    .str.strip("_")
)


unnamed_columns = [
    column
    for column in emergency_df.columns
    if column.startswith("unnamed")
]


emergency_df = emergency_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)


text_candidates = [
    "text",
    "message",
    "complaint_text",
    "complaint_description",
    "description",
    "interaction_summary",
    "emergency_text",
    "incident_description",
    "feedback_text",
    "content",
    "summary"
]


target_candidates = [
    "emergency",
    "is_emergency",
    "emergency_label",
    "emergency_status",
    "label",
    "target",
    "class",
    "incident_type"
]


text_column = next(
    (
        column
        for column in text_candidates
        if column in emergency_df.columns
    ),
    None
)


target_column = next(
    (
        column
        for column in target_candidates
        if column in emergency_df.columns
    ),
    None
)


if text_column is None:

    raise ValueError(
        "No emergency text column was found. "
        f"Available columns: "
        f"{emergency_df.columns.tolist()}"
    )


if target_column is None:

    raise ValueError(
        "No emergency target column was found. "
        f"Available columns: "
        f"{emergency_df.columns.tolist()}"
    )


print("\nInput text column:", text_column)
print("Target emergency column:", target_column)


emergency_df[text_column] = (
    emergency_df[text_column]
    .fillna("")
    .astype(str)
    .str.strip()
)


emergency_df[target_column] = (
    emergency_df[target_column]
    .fillna("")
    .astype(str)
    .str.strip()
)


emergency_df["emergency_text"] = (
    emergency_df[text_column]
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
    .str.strip()
)


def standardize_emergency_label(value):

    value = str(value).strip().lower()

    emergency_values = [
        "1",
        "yes",
        "true",
        "emergency",
        "urgent",
        "critical",
        "high risk",
        "high",
        "positive"
    ]

    non_emergency_values = [
        "0",
        "no",
        "false",
        "non emergency",
        "non-emergency",
        "normal",
        "not emergency",
        "low risk",
        "negative"
    ]

    if value in emergency_values:
        return "Emergency"

    if value in non_emergency_values:
        return "Non-Emergency"

    return str(value).strip().title()


emergency_df["emergency_label"] = (
    emergency_df[target_column]
    .apply(standardize_emergency_label)
)


def remove_emergency_label_leakage(value):

    value = str(value)

    value = re.sub(
        r"\bemergency\s*[:\-]?\s*"
        r"(yes|no|true|false|1|0)\b",
        " ",
        value,
        flags=re.IGNORECASE
    )

    value = re.sub(
        r"\bis[_\s-]*emergency\s*[:\-]?\s*"
        r"(yes|no|true|false|1|0)\b",
        " ",
        value,
        flags=re.IGNORECASE
    )

    value = re.sub(
        r"\blabel\s*[:\-]?\s*"
        r"(emergency|non[-\s]?emergency)\b",
        " ",
        value,
        flags=re.IGNORECASE
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


emergency_df["emergency_text"] = (
    emergency_df["emergency_text"]
    .apply(remove_emergency_label_leakage)
)


def clean_emergency_text(value):

    value = str(value).lower()

    value = re.sub(
        r"http\S+|www\S+",
        " ",
        value
    )

    value = re.sub(
        r"\S+@\S+",
        " ",
        value
    )

    value = re.sub(
        r"@\w+",
        " ",
        value
    )

    value = re.sub(
        r"#(\w+)",
        r"\1",
        value
    )

    value = re.sub(
        r"&[a-z]+;",
        " ",
        value
    )

    value = re.sub(
        r"[^a-z\s]",
        " ",
        value
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


emergency_df["clean_text"] = (
    emergency_df["emergency_text"]
    .apply(clean_emergency_text)
)


selected_columns = [
    "emergency_text",
    "clean_text",
    "emergency_label"
]


optional_columns = [
    "incident_id",
    "complaint_id",
    "interaction_id",
    "citizen_id",
    "timestamp",
    "date",
    "location",
    "state",
    "district",
    "department",
    "priority",
    "category",
    "channel",
    "language",
    "sentiment",
    "severity",
    "response_time",
    "resolved"
]


for column in optional_columns:

    if column in emergency_df.columns:
        selected_columns.append(column)


model_df = emergency_df[
    selected_columns
].copy()


model_df = model_df.dropna(
    subset=[
        "clean_text",
        "emergency_label"
    ]
)


invalid_labels = [
    "",
    "nan",
    "none",
    "null",
    "n/a",
    "na",
    "not available",
    "unknown",
    "-"
]


model_df = model_df[
    model_df["clean_text"].str.len() >= 5
].copy()


model_df = model_df[
    ~model_df["emergency_label"]
    .str.lower()
    .isin(invalid_labels)
].copy()


model_df = model_df.drop_duplicates(
    subset=[
        "clean_text",
        "emergency_label"
    ]
).reset_index(drop=True)


emergency_counts_before = (
    model_df["emergency_label"]
    .value_counts()
)


print("\nEmergency distribution before filtering:")

display(
    emergency_counts_before
    .rename_axis("emergency_status")
    .reset_index(name="record_count")
)


minimum_records_per_class = 10


valid_classes = emergency_counts_before[
    emergency_counts_before
    >= minimum_records_per_class
].index


model_df = model_df[
    model_df["emergency_label"].isin(
        valid_classes
    )
].reset_index(drop=True)


if len(model_df) < 20:

    raise ValueError(
        "Not enough usable records are available."
    )


if model_df["emergency_label"].nunique() < 2:

    raise ValueError(
        "At least two emergency classes are required."
    )


minimum_class_count = (
    model_df["emergency_label"]
    .value_counts()
    .min()
)


if minimum_class_count < 2:

    raise ValueError(
        "Each emergency class must contain "
        "at least two records."
    )


print("\nFinal emergency distribution:")

display(
    model_df["emergency_label"]
    .value_counts()
    .rename_axis("emergency_status")
    .reset_index(name="record_count")
)


print("\nOriginal records:", len(emergency_df))
print("Usable records:", len(model_df))

print(
    "Number of emergency classes:",
    model_df["emergency_label"].nunique()
)


X = model_df["clean_text"]

y = model_df["emergency_label"]


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))


emergency_detection_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            stop_words="english",
            max_features=15000,
            ngram_range=(1, 2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            token_pattern=(
                r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
            )
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            solver="liblinear",
            random_state=42
        )
    )
])


emergency_detection_model.fit(
    X_train,
    y_train
)


print(
    "\nEmergency detection model trained successfully."
)


y_pred = emergency_detection_model.predict(
    X_test
)


prediction_probabilities = (
    emergency_detection_model.predict_proba(
        X_test
    )
)


prediction_confidence = (
    prediction_probabilities.max(axis=1)
)


accuracy = accuracy_score(
    y_test,
    y_pred
)


weighted_precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


weighted_recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)


print("\nEmergency Detection Model Evaluation")


print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


print(
    "Weighted precision:",
    round(weighted_precision * 100, 2),
    "%"
)


print(
    "Weighted recall:",
    round(weighted_recall * 100, 2),
    "%"
)


print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)


print(
    "Macro F1-score:",
    round(macro_f1 * 100, 2),
    "%"
)


print("\nClassification report:")


print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


report_df = pd.DataFrame(
    classification_report(
        y_test,
        y_pred,
        output_dict=True,
        zero_division=0
    )
).transpose().reset_index()


report_df = report_df.rename(
    columns={
        "index": "class_or_metric"
    }
)


display(report_df)


class_labels = sorted(
    set(y_test.unique()).union(
        set(y_pred)
    )
)


matrix = confusion_matrix(
    y_test,
    y_pred,
    labels=class_labels
)


confusion_matrix_df = pd.DataFrame(
    matrix,
    index=class_labels,
    columns=class_labels
)


print("\nConfusion matrix:")

display(confusion_matrix_df)


fig, ax = plt.subplots(
    figsize=(8, 7)
)


display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_labels
)


display_matrix.plot(
    ax=ax,
    values_format="d",
    xticks_rotation=45
)


plt.title(
    "Emergency Detection Confusion Matrix"
)


plt.tight_layout()
plt.show()


prediction_results = pd.DataFrame({
    "emergency_text": X_test.values,
    "actual_emergency_status": y_test.values,
    "predicted_emergency_status": y_pred,
    "prediction_confidence": prediction_confidence
})


prediction_results[
    "prediction_confidence_percentage"
] = (
    prediction_results[
        "prediction_confidence"
    ] * 100
).round(2)


prediction_results[
    "prediction_correct"
] = (
    prediction_results[
        "actual_emergency_status"
    ]
    == prediction_results[
        "predicted_emergency_status"
    ]
)


prediction_results.insert(
    0,
    "prediction_id",
    range(
        1,
        len(prediction_results) + 1
    )
)


print("\nEmergency detection results:")

display(
    prediction_results.head(20)
)


evaluation_df = pd.DataFrame({
    "metric": [
        "accuracy",
        "weighted_precision",
        "weighted_recall",
        "weighted_f1_score",
        "macro_f1_score",
        "original_records",
        "usable_records",
        "training_records",
        "testing_records",
        "number_of_emergency_classes",
        "minimum_records_per_class"
    ],
    "value": [
        float(accuracy),
        float(weighted_precision),
        float(weighted_recall),
        float(weighted_f1),
        float(macro_f1),
        int(len(emergency_df)),
        int(len(model_df)),
        int(len(X_train)),
        int(len(X_test)),
        int(y.nunique()),
        int(minimum_records_per_class)
    ]
})


print("\nModel evaluation summary:")

display(evaluation_df)


labelled_data = model_df.copy()


labelled_data.insert(
    0,
    "labelled_record_id",
    range(
        1,
        len(labelled_data) + 1
    )
)


distribution_df = (
    model_df["emergency_label"]
    .value_counts()
    .rename_axis("emergency_status")
    .reset_index(name="record_count")
)


confusion_sql_df = (
    confusion_matrix_df
    .reset_index()
    .rename(
        columns={
            "index": "actual_emergency_status"
        }
    )
)


model_filename = (
    "06_emergency_detection_model.pkl"
)


joblib.dump(
    emergency_detection_model,
    model_filename
)


print(
    "\nModel saved successfully:",
    model_filename
)


prediction_results.to_csv(
    "06_emergency_detection_results.csv",
    index=False
)


evaluation_df.to_csv(
    "06_emergency_model_evaluation.csv",
    index=False
)


report_df.to_csv(
    "06_emergency_classification_report.csv",
    index=False
)


labelled_data.to_csv(
    "06_emergency_labelled_data.csv",
    index=False
)


distribution_df.to_csv(
    "06_emergency_class_distribution.csv",
    index=False
)


confusion_sql_df.to_csv(
    "06_emergency_confusion_matrix.csv",
    index=False
)


print("\nCSV files saved successfully.")


prediction_results.to_sql(
    name="emergency_detection_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)


evaluation_df.to_sql(
    name="emergency_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)


report_df.to_sql(
    name="emergency_classification_report",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)


labelled_data.to_sql(
    name="emergency_labelled_data",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)


distribution_df.to_sql(
    name="emergency_class_distribution",
    con=database_engine,
    if_exists="replace",
    index=False
)


confusion_sql_df.to_sql(
    name="emergency_confusion_matrix",
    con=database_engine,
    if_exists="replace",
    index=False
)


print(
    "\nAll Emergency Detection tables created in MySQL."
)


def predict_emergency(new_message):

    if not isinstance(new_message, str):

        raise TypeError(
            "The emergency message must be entered as text."
        )

    cleaned_message = remove_emergency_label_leakage(
        new_message
    )

    cleaned_message = clean_emergency_text(
        cleaned_message
    )

    if len(cleaned_message) < 3:

        raise ValueError(
            "Enter a meaningful emergency message."
        )

    predicted_status = (
        emergency_detection_model.predict(
            [cleaned_message]
        )[0]
    )

    probabilities = (
        emergency_detection_model.predict_proba(
            [cleaned_message]
        )[0]
    )

    probability_df = pd.DataFrame({
        "emergency_status":
            emergency_detection_model.classes_,
        "probability_percentage": (
            probabilities * 100
        ).round(2)
    })

    probability_df = (
        probability_df
        .sort_values(
            by="probability_percentage",
            ascending=False
        )
        .reset_index(drop=True)
    )

    confidence = float(
        probability_df.iloc[0][
            "probability_percentage"
        ]
    )

    print("\nMessage:", new_message)

    print(
        "Predicted emergency status:",
        predicted_status
    )

    print(
        "Prediction confidence:",
        confidence,
        "%"
    )

    print("\nEmergency probabilities:")

    display(probability_df)

    return {
        "message": new_message,
        "predicted_emergency_status": predicted_status,
        "confidence_percentage": confidence
    }


predict_emergency(
    "A person has collapsed and is not breathing. "
    "An ambulance is required immediately."
)


predict_emergency(
    "There is a large fire in a residential building "
    "and people are trapped inside."
)


predict_emergency(
    "I want to know the status of my government "
    "scheme application."
)


predict_emergency(
    "A major road accident has occurred and several "
    "people are seriously injured."
)


predict_emergency(
    "The streetlight near my house has not been "
    "working for two days."
)


verification_df = pd.DataFrame({
    "table_name": [
        "emergency_detection_results",
        "emergency_model_evaluation",
        "emergency_classification_report",
        "emergency_labelled_data",
        "emergency_class_distribution",
        "emergency_confusion_matrix"
    ],
    "total_rows": [
        len(prediction_results),
        len(evaluation_df),
        len(report_df),
        len(labelled_data),
        len(distribution_df),
        len(confusion_sql_df)
    ]
})


print("\nMySQL tables created:")

display(verification_df)


sql_prediction_sample = pd.read_sql(
    """
    SELECT *
    FROM emergency_detection_results
    LIMIT 10
    """,
    database_engine
)


print("\nSample predictions read from MySQL:")

display(sql_prediction_sample)


print(
    "\n06_Emergency_Detection completed successfully."
)


print("\nSource table:", source_table)
print("Input column:", text_column)
print("Target column:", target_column)
print("Original records:", len(emergency_df))
print("Usable records:", len(model_df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print(
    "Number of emergency classes:",
    y.nunique()
)


print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)


print("\nSaved model:", model_filename)


print("\nMySQL output tables:")

print("1. emergency_detection_results")
print("2. emergency_model_evaluation")
print("3. emergency_classification_report")
print("4. emergency_labelled_data")
print("5. emergency_class_distribution")
print("6. emergency_confusion_matrix")
```
