```python
import os
import re
import getpass
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"

database_name = "government_analytics"
source_table = "government_action_recommendation"

test_size = 0.20
random_state = 42
minimum_records_required = 2

mysql_password = getpass.getpass("Enter MySQL password: ")
encoded_password = quote_plus(mysql_password)

server_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)

with server_engine.begin() as connection:
    connection.execute(
        text(f"CREATE DATABASE IF NOT EXISTS `{database_name}`")
    )

database_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/{database_name}",
    pool_pre_ping=True
)

with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()

print("Connected database:", connected_database)

existing_tables = inspect(database_engine).get_table_names()

possible_paths = [
    "government_action_recommendation.csv",
    "government_action_recommendation(1).csv",
    "government_action_recommendation(2).csv",
    "10_Government_Action_Recommendation.csv",
    "citizen_complaints.csv",
    "final_citizen_complaints.csv",
    "government_analytics_data.csv",
    "unified_training_dataset.csv",
    "09_anomaly_detection_results.csv",
    "08_trend_forecasting_test_results.csv",
    "/mnt/data/government_action_recommendation.csv",
    "/mnt/data/government_action_recommendation(1).csv",
    "/mnt/data/government_action_recommendation(2).csv",
    "/mnt/data/10_Government_Action_Recommendation.csv",
    "/mnt/data/citizen_complaints.csv",
    "/mnt/data/final_citizen_complaints.csv",
    "/mnt/data/government_analytics_data.csv",
    "/mnt/data/unified_training_dataset.csv",
    "/mnt/data/09_anomaly_detection_results.csv",
    r"C:\Users\Kishore\Downloads\government_action_recommendation.csv",
    r"C:\Users\Kishore\Downloads\government_action_recommendation(1).csv",
    r"C:\Users\Kishore\Downloads\government_action_recommendation(2).csv",
    r"C:\Users\Kishore\Downloads\10_Government_Action_Recommendation.csv",
    r"C:\Users\Kishore\Downloads\citizen_complaints.csv",
    r"C:\Users\Kishore\Downloads\final_citizen_complaints.csv",
    r"C:\Users\Kishore\Downloads\government_analytics_data.csv"
]

csv_path = next(
    (path for path in possible_paths if os.path.exists(path)),
    None
)

if csv_path is not None:
    government_df = pd.read_csv(csv_path, low_memory=False)

    government_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print("Dataset loaded from CSV:", csv_path)
    print("Dataset uploaded to MySQL table:", source_table)

elif source_table in existing_tables:
    government_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print("Dataset loaded from MySQL table:", source_table)

else:
    alternative_tables = [
        "anomaly_detection_results",
        "priority_prediction_results",
        "emergency_detection_results",
        "harmful_content_detection_results",
        "complaint_reason_classification_results",
        "citizen_complaints",
        "government_complaints",
        "complaints",
        "unified_training_dataset",
        "government_social_media"
    ]

    selected_table = next(
        (
            table_name
            for table_name in alternative_tables
            if table_name in existing_tables
        ),
        None
    )

    if selected_table is None:
        raise FileNotFoundError(
            "Place government_action_recommendation.csv, "
            "10_Government_Action_Recommendation.csv, "
            "citizen_complaints.csv or final_citizen_complaints.csv "
            "in the same folder as the notebook."
        )

    government_df = pd.read_sql(
        f"SELECT * FROM `{selected_table}`",
        database_engine
    )

    source_table = selected_table
    print("Dataset loaded from alternative MySQL table:", source_table)

if government_df.empty:
    raise ValueError("The government action recommendation dataset is empty.")

print("\nOriginal dataset shape:", government_df.shape)
print("\nOriginal columns:")
print(government_df.columns.tolist())
display(government_df.head())

government_df.columns = (
    government_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

unnamed_columns = [
    column
    for column in government_df.columns
    if column.startswith("unnamed")
]

government_df = government_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)

print("\nStandardized columns:")
print(government_df.columns.tolist())

text_candidates = [
    "complaint_text",
    "complaint_description",
    "description",
    "complaint",
    "text",
    "message",
    "content",
    "post_text",
    "tweet_text",
    "feedback",
    "citizen_feedback",
    "issue_description",
    "grievance",
    "reason",
    "subject"
]

available_text_columns = [
    column
    for column in text_candidates
    if column in government_df.columns
]

if not available_text_columns:
    object_columns = government_df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    excluded_text_keywords = [
        "id",
        "date",
        "time",
        "label",
        "target",
        "action",
        "recommendation",
        "status",
        "priority",
        "severity",
        "sentiment",
        "category",
        "department",
        "district",
        "state",
        "platform",
        "channel"
    ]

    available_text_columns = [
        column
        for column in object_columns
        if not any(
            keyword in column.lower()
            for keyword in excluded_text_keywords
        )
    ]

if not available_text_columns:
    raise ValueError(
        "No usable complaint or message text column was found."
    )

selected_text_columns = available_text_columns[:3]

government_df["combined_text"] = (
    government_df[selected_text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

government_df = government_df[
    government_df["combined_text"].str.len() >= 3
].reset_index(drop=True)

if len(government_df) < minimum_records_required:
    raise ValueError(
        "At least two usable text records are required."
    )

print("\nSelected text columns:", selected_text_columns)
print("Usable text records:", len(government_df))

label_candidates = [
    "recommended_action",
    "government_action",
    "action_recommendation",
    "recommended_government_action",
    "action",
    "recommendation",
    "resolution_action",
    "suggested_action",
    "target",
    "label"
]

target_column = next(
    (
        column
        for column in label_candidates
        if column in government_df.columns
    ),
    None
)

def normalize_text_value(value):
    if pd.isna(value):
        return ""
    return re.sub(r"\s+", " ", str(value).strip())

def normalize_label(value):
    value = normalize_text_value(value).lower()

    mapping = {
        "immediate emergency response": "Immediate Emergency Response",
        "emergency response": "Immediate Emergency Response",
        "dispatch emergency services": "Immediate Emergency Response",
        "police action": "Law Enforcement Action",
        "law enforcement": "Law Enforcement Action",
        "legal action": "Law Enforcement Action",
        "cyber crime investigation": "Cybercrime Investigation",
        "cybercrime investigation": "Cybercrime Investigation",
        "content removal": "Content Moderation and Removal",
        "remove harmful content": "Content Moderation and Removal",
        "public works repair": "Public Infrastructure Repair",
        "infrastructure repair": "Public Infrastructure Repair",
        "road repair": "Public Infrastructure Repair",
        "water supply action": "Water and Sanitation Action",
        "water and sanitation": "Water and Sanitation Action",
        "sanitation action": "Water and Sanitation Action",
        "health inspection": "Public Health Intervention",
        "public health action": "Public Health Intervention",
        "welfare support": "Social Welfare Assistance",
        "social welfare": "Social Welfare Assistance",
        "financial investigation": "Financial and Fraud Investigation",
        "fraud investigation": "Financial and Fraud Investigation",
        "consumer protection": "Consumer Protection Action",
        "department review": "Departmental Review and Resolution",
        "administrative review": "Departmental Review and Resolution",
        "monitor": "Monitoring and Follow-up",
        "follow up": "Monitoring and Follow-up",
        "awareness campaign": "Public Awareness Campaign",
        "no action": "No Immediate Action Required"
    }

    if value in mapping:
        return mapping[value]

    if value == "":
        return ""

    return value.title()

def generate_action_from_text(text_value):
    text_value = normalize_text_value(text_value).lower()

    emergency_terms = [
        "fire", "accident", "injured", "injury", "ambulance",
        "medical emergency", "life threat", "collapsed",
        "explosion", "flood", "rescue", "missing person"
    ]

    law_terms = [
        "murder", "attack", "assault", "violence", "threat",
        "harassment", "theft", "robbery", "kidnap", "crime",
        "illegal weapon", "domestic violence"
    ]

    cyber_terms = [
        "cyber", "online fraud", "hacked", "hacking", "phishing",
        "otp fraud", "account hacked", "identity theft",
        "fake account", "data breach"
    ]

    harmful_content_terms = [
        "hate speech", "abusive post", "harmful content",
        "fake news", "misinformation", "obscene", "pornographic",
        "violent content", "terror content", "remove post"
    ]

    infrastructure_terms = [
        "pothole", "road damaged", "broken road", "streetlight",
        "street light", "bridge", "drainage", "electric pole",
        "power line", "public building"
    ]

    water_terms = [
        "water shortage", "no water", "dirty water", "sewage",
        "garbage", "waste", "sanitation", "drain overflow",
        "contaminated water", "toilet"
    ]

    health_terms = [
        "hospital", "disease", "food poisoning", "mosquito",
        "dengue", "malaria", "unsafe food", "health hazard",
        "medicine shortage", "epidemic"
    ]

    welfare_terms = [
        "pension", "ration", "scholarship", "benefit",
        "welfare scheme", "subsidy", "housing scheme",
        "disabled support", "widow pension", "farmer support"
    ]

    financial_terms = [
        "bribe", "corruption", "money laundering",
        "financial fraud", "bank fraud", "embezzlement",
        "fund misuse", "fake payment"
    ]

    consumer_terms = [
        "overcharged", "defective product", "refund",
        "consumer complaint", "false advertisement",
        "service charge", "billing issue"
    ]

    awareness_terms = [
        "awareness", "education campaign", "public information",
        "safety campaign", "training programme"
    ]

    monitoring_terms = [
        "delay", "pending", "not resolved", "no response",
        "follow up", "status update", "repeated complaint"
    ]

    if any(term in text_value for term in emergency_terms):
        return "Immediate Emergency Response"

    if any(term in text_value for term in law_terms):
        return "Law Enforcement Action"

    if any(term in text_value for term in cyber_terms):
        return "Cybercrime Investigation"

    if any(term in text_value for term in harmful_content_terms):
        return "Content Moderation and Removal"

    if any(term in text_value for term in infrastructure_terms):
        return "Public Infrastructure Repair"

    if any(term in text_value for term in water_terms):
        return "Water and Sanitation Action"

    if any(term in text_value for term in health_terms):
        return "Public Health Intervention"

    if any(term in text_value for term in welfare_terms):
        return "Social Welfare Assistance"

    if any(term in text_value for term in financial_terms):
        return "Financial and Fraud Investigation"

    if any(term in text_value for term in consumer_terms):
        return "Consumer Protection Action"

    if any(term in text_value for term in awareness_terms):
        return "Public Awareness Campaign"

    if any(term in text_value for term in monitoring_terms):
        return "Monitoring and Follow-up"

    return "Departmental Review and Resolution"

if target_column is not None:
    government_df["recommended_action"] = (
        government_df[target_column]
        .apply(normalize_label)
    )

    missing_target_mask = (
        government_df["recommended_action"].eq("")
    )

    government_df.loc[
        missing_target_mask,
        "recommended_action"
    ] = government_df.loc[
        missing_target_mask,
        "combined_text"
    ].apply(generate_action_from_text)

    print("\nExisting target column detected:", target_column)

else:
    government_df["recommended_action"] = (
        government_df["combined_text"]
        .apply(generate_action_from_text)
    )

    target_column = "generated_recommended_action"
    print("\nNo existing action label was found.")
    print("Action labels were generated using government-response rules.")

government_df["recommended_action"] = (
    government_df["recommended_action"]
    .fillna("Departmental Review and Resolution")
    .astype(str)
    .str.strip()
)

label_distribution_df = (
    government_df["recommended_action"]
    .value_counts()
    .rename_axis("recommended_action")
    .reset_index(name="record_count")
)

label_distribution_df["percentage"] = (
    label_distribution_df["record_count"]
    / len(government_df)
    * 100
).round(2)

print("\nAction distribution:")
display(label_distribution_df)

X = government_df["combined_text"].copy()
y = government_df["recommended_action"].copy()

class_counts = y.value_counts()
usable_classes = class_counts[class_counts >= 1].index.tolist()

training_df = government_df[
    government_df["recommended_action"].isin(usable_classes)
].copy().reset_index(drop=True)

X = training_df["combined_text"]
y = training_df["recommended_action"]

unique_class_count = y.nunique()
model_mode = "rule_based"
recommendation_model = None
evaluation_available = False
accuracy = np.nan
weighted_precision = np.nan
weighted_recall = np.nan
weighted_f1 = np.nan
report_df = pd.DataFrame()
confusion_matrix_df = pd.DataFrame()

if unique_class_count >= 2 and len(training_df) >= 4:
    minimum_class_count = y.value_counts().min()
    stratify_value = y if minimum_class_count >= 2 else None

    calculated_test_size = max(
        int(round(len(training_df) * test_size)),
        unique_class_count
    )

    calculated_test_size = min(
        calculated_test_size,
        len(training_df) - 1
    )

    if calculated_test_size >= 1:
        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=calculated_test_size,
            random_state=random_state,
            stratify=stratify_value
        )

        recommendation_model = Pipeline([
            (
                "tfidf",
                TfidfVectorizer(
                    lowercase=True,
                    stop_words="english",
                    ngram_range=(1, 2),
                    min_df=1,
                    max_df=1.0,
                    max_features=30000,
                    sublinear_tf=True
                )
            ),
            (
                "classifier",
                LogisticRegression(
                    max_iter=2000,
                    class_weight="balanced",
                    random_state=random_state
                )
            )
        ])

        recommendation_model.fit(X_train, y_train)
        y_pred = recommendation_model.predict(X_test)

        evaluation_available = True
        model_mode = "machine_learning"

        accuracy = accuracy_score(y_test, y_pred)
        weighted_precision = precision_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )
        weighted_recall = recall_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )
        weighted_f1 = f1_score(
            y_test,
            y_pred,
            average="weighted",
            zero_division=0
        )

        print("\nGovernment Action Recommendation Evaluation")
        print("Accuracy:", round(accuracy * 100, 2), "%")
        print("Weighted Precision:", round(weighted_precision * 100, 2), "%")
        print("Weighted Recall:", round(weighted_recall * 100, 2), "%")
        print("Weighted F1-score:", round(weighted_f1 * 100, 2), "%")

        print("\nClassification report:")
        print(
            classification_report(
                y_test,
                y_pred,
                zero_division=0
            )
        )

        report_df = pd.DataFrame(
            classification_report(
                y_test,
                y_pred,
                output_dict=True,
                zero_division=0
            )
        ).transpose().reset_index()

        report_df = report_df.rename(
            columns={"index": "class_or_metric"}
        )

        class_labels = sorted(y.unique().tolist())

        matrix = confusion_matrix(
            y_test,
            y_pred,
            labels=class_labels
        )

        confusion_matrix_df = pd.DataFrame(
            matrix,
            index=class_labels,
            columns=class_labels
        )

        fig, ax = plt.subplots(figsize=(12, 10))

        matrix_display = ConfusionMatrixDisplay(
            confusion_matrix=matrix,
            display_labels=class_labels
        )

        matrix_display.plot(
            ax=ax,
            values_format="d",
            xticks_rotation=45
        )

        plt.title("Government Action Recommendation Confusion Matrix")
        plt.tight_layout()
        plt.show()

    else:
        print("\nThe dataset is too small for a train/test split.")
        print("Rule-based recommendation mode will be used.")

else:
    print("\nThe dataset does not contain enough classes or records")
    print("for machine-learning training.")
    print("Rule-based recommendation mode will be used.")

if recommendation_model is not None:
    government_df["predicted_action"] = recommendation_model.predict(
        government_df["combined_text"]
    )

    probabilities = recommendation_model.predict_proba(
        government_df["combined_text"]
    )

    government_df["recommendation_confidence"] = (
        probabilities.max(axis=1) * 100
    ).round(2)

else:
    government_df["predicted_action"] = (
        government_df["combined_text"]
        .apply(generate_action_from_text)
    )

    government_df["recommendation_confidence"] = 100.0

government_df["action_match"] = np.where(
    government_df["recommended_action"]
    == government_df["predicted_action"],
    "Matched",
    "Different"
)

priority_candidates = [
    "predicted_priority",
    "priority",
    "priority_label",
    "priority_level"
]

priority_column = next(
    (
        column
        for column in priority_candidates
        if column in government_df.columns
    ),
    None
)

emergency_candidates = [
    "predicted_emergency_status",
    "emergency_status",
    "emergency_label",
    "is_emergency"
]

emergency_column = next(
    (
        column
        for column in emergency_candidates
        if column in government_df.columns
    ),
    None
)

harmful_candidates = [
    "predicted_harmful_status",
    "harmful_status",
    "harmful_label",
    "is_harmful"
]

harmful_column = next(
    (
        column
        for column in harmful_candidates
        if column in government_df.columns
    ),
    None
)

anomaly_candidates = [
    "anomaly_status",
    "predicted_anomaly_status",
    "is_anomaly"
]

anomaly_column = next(
    (
        column
        for column in anomaly_candidates
        if column in government_df.columns
    ),
    None
)

def determine_action_urgency(row):
    action = normalize_text_value(
        row.get("predicted_action", "")
    ).lower()

    priority = normalize_text_value(
        row.get(priority_column, "")
        if priority_column is not None
        else ""
    ).lower()

    emergency = normalize_text_value(
        row.get(emergency_column, "")
        if emergency_column is not None
        else ""
    ).lower()

    harmful = normalize_text_value(
        row.get(harmful_column, "")
        if harmful_column is not None
        else ""
    ).lower()

    anomaly = normalize_text_value(
        row.get(anomaly_column, "")
        if anomaly_column is not None
        else ""
    ).lower()

    if (
        "immediate emergency" in action
        or emergency in {"emergency", "yes", "true", "1", "1.0"}
        or priority in {"critical", "urgent", "high"}
    ):
        return "Immediate"

    if (
        "law enforcement" in action
        or "cybercrime" in action
        or "content moderation" in action
        or harmful in {"harmful", "yes", "true", "1", "1.0"}
        or anomaly in {"anomaly", "yes", "true", "1", "1.0"}
    ):
        return "Within 24 Hours"

    if (
        "public infrastructure" in action
        or "water and sanitation" in action
        or "public health" in action
        or "financial and fraud" in action
    ):
        return "Within 3 Days"

    if (
        "social welfare" in action
        or "consumer protection" in action
        or "monitoring" in action
    ):
        return "Within 7 Days"

    return "Standard Department Timeline"

government_df["action_urgency"] = government_df.apply(
    determine_action_urgency,
    axis=1
)

department_mapping = {
    "Immediate Emergency Response": "Emergency Services",
    "Law Enforcement Action": "Police Department",
    "Cybercrime Investigation": "Cyber Crime Department",
    "Content Moderation and Removal": "Information Technology and Media Cell",
    "Public Infrastructure Repair": "Public Works Department",
    "Water and Sanitation Action": "Municipal and Water Department",
    "Public Health Intervention": "Health Department",
    "Social Welfare Assistance": "Social Welfare Department",
    "Financial and Fraud Investigation": "Vigilance and Finance Department",
    "Consumer Protection Action": "Consumer Affairs Department",
    "Public Awareness Campaign": "Public Relations Department",
    "Monitoring and Follow-up": "Grievance Monitoring Cell",
    "Departmental Review and Resolution": "Concerned Administrative Department",
    "No Immediate Action Required": "Grievance Monitoring Cell"
}

government_df["responsible_department"] = (
    government_df["predicted_action"]
    .map(department_mapping)
    .fillna("Concerned Administrative Department")
)

government_df.insert(
    0,
    "recommendation_record_id",
    range(1, len(government_df) + 1)
)

action_summary_df = (
    government_df.groupby(
        [
            "predicted_action",
            "responsible_department",
            "action_urgency"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="record_count")
    .sort_values(
        by="record_count",
        ascending=False
    )
    .reset_index(drop=True)
)

action_summary_df["percentage"] = (
    action_summary_df["record_count"]
    / len(government_df)
    * 100
).round(2)

print("\nGovernment action summary:")
display(action_summary_df)

plt.figure(figsize=(12, 7))
plot_action_counts = (
    government_df["predicted_action"]
    .value_counts()
    .sort_values(ascending=True)
)
plt.barh(
    plot_action_counts.index,
    plot_action_counts.values
)
plt.title("10 Government Action Recommendation Distribution")
plt.xlabel("Number of Records")
plt.ylabel("Recommended Government Action")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
urgency_counts = (
    government_df["action_urgency"]
    .value_counts()
)
plt.bar(
    urgency_counts.index,
    urgency_counts.values
)
plt.title("Government Action Urgency Distribution")
plt.xlabel("Action Urgency")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

evaluation_df = pd.DataFrame({
    "metric": [
        "model_mode",
        "source_table",
        "selected_text_columns",
        "target_column",
        "total_records",
        "number_of_action_classes",
        "evaluation_available",
        "accuracy",
        "weighted_precision",
        "weighted_recall",
        "weighted_f1_score"
    ],
    "value": [
        model_mode,
        source_table,
        ", ".join(selected_text_columns),
        target_column,
        str(len(government_df)),
        str(government_df["recommended_action"].nunique()),
        str(evaluation_available),
        (
            str(float(accuracy))
            if pd.notna(accuracy)
            else "Not Available"
        ),
        (
            str(float(weighted_precision))
            if pd.notna(weighted_precision)
            else "Not Available"
        ),
        (
            str(float(weighted_recall))
            if pd.notna(weighted_recall)
            else "Not Available"
        ),
        (
            str(float(weighted_f1))
            if pd.notna(weighted_f1)
            else "Not Available"
        )
    ]
})

print("\nModel evaluation summary:")
display(evaluation_df)

result_columns = [
    "recommendation_record_id"
]

identifier_candidates = [
    "complaint_id",
    "record_id",
    "interaction_id",
    "post_id",
    "user_id",
    "citizen_id",
    "department",
    "category",
    "complaint_category",
    "district",
    "state",
    "platform",
    "channel",
    "priority",
    "predicted_priority",
    "emergency_status",
    "predicted_emergency_status",
    "harmful_status",
    "predicted_harmful_status",
    "anomaly_status",
    "date",
    "complaint_date",
    "created_at"
]

for column in identifier_candidates:
    if (
        column in government_df.columns
        and column not in result_columns
    ):
        result_columns.append(column)

result_columns.extend([
    "combined_text",
    "recommended_action",
    "predicted_action",
    "recommendation_confidence",
    "action_match",
    "responsible_department",
    "action_urgency"
])

government_action_results_df = government_df[
    result_columns
].copy()

print("\nFinal government action recommendations:")
display(government_action_results_df.head(20))

model_package = {
    "model": recommendation_model,
    "model_mode": model_mode,
    "selected_text_columns": selected_text_columns,
    "target_column": target_column,
    "department_mapping": department_mapping,
    "rule_function_name": "generate_action_from_text",
    "action_classes": sorted(
        government_df["predicted_action"].unique().tolist()
    )
}

model_filename = "10_government_action_recommendation_model.pkl"

joblib.dump(
    model_package,
    model_filename
)

print("\nModel package saved successfully:", model_filename)

government_action_results_df.to_csv(
    "10_government_action_recommendation_results.csv",
    index=False
)

action_summary_df.to_csv(
    "10_government_action_summary.csv",
    index=False
)

evaluation_df.to_csv(
    "10_government_action_model_evaluation.csv",
    index=False
)

label_distribution_df.to_csv(
    "10_government_action_label_distribution.csv",
    index=False
)

if not report_df.empty:
    report_df.to_csv(
        "10_government_action_classification_report.csv",
        index=False
    )

if not confusion_matrix_df.empty:
    confusion_sql_df = (
        confusion_matrix_df
        .reset_index()
        .rename(columns={"index": "actual_action"})
    )

    confusion_sql_df.to_csv(
        "10_government_action_confusion_matrix.csv",
        index=False
    )
else:
    confusion_sql_df = pd.DataFrame()

print("\nCSV files saved successfully.")

government_action_results_df.to_sql(
    name="government_action_recommendation_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

action_summary_df.to_sql(
    name="government_action_summary",
    con=database_engine,
    if_exists="replace",
    index=False
)

evaluation_df.to_sql(
    name="government_action_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)

label_distribution_df.to_sql(
    name="government_action_label_distribution",
    con=database_engine,
    if_exists="replace",
    index=False
)

if not report_df.empty:
    report_df.to_sql(
        name="government_action_classification_report",
        con=database_engine,
        if_exists="replace",
        index=False
    )

if not confusion_sql_df.empty:
    confusion_sql_df.to_sql(
        name="government_action_confusion_matrix",
        con=database_engine,
        if_exists="replace",
        index=False
    )

print("\nAll Government Action Recommendation tables created in MySQL.")

def recommend_government_action(
    complaint_text,
    priority=None,
    emergency_status=None,
    harmful_status=None,
    anomaly_status=None
):
    complaint_text = normalize_text_value(complaint_text)

    if complaint_text == "":
        raise ValueError("Complaint text cannot be empty.")

    if recommendation_model is not None:
        predicted_action = recommendation_model.predict(
            [complaint_text]
        )[0]

        probability_values = recommendation_model.predict_proba(
            [complaint_text]
        )[0]

        confidence = float(
            probability_values.max() * 100
        )

    else:
        predicted_action = generate_action_from_text(
            complaint_text
        )
        confidence = 100.0

    temporary_row = pd.Series({
        "predicted_action": predicted_action,
        priority_column if priority_column is not None else "priority": priority,
        emergency_column if emergency_column is not None else "emergency_status": emergency_status,
        harmful_column if harmful_column is not None else "harmful_status": harmful_status,
        anomaly_column if anomaly_column is not None else "anomaly_status": anomaly_status
    })

    action_urgency = determine_action_urgency(
        temporary_row
    )

    responsible_department = department_mapping.get(
        predicted_action,
        "Concerned Administrative Department"
    )

    result = {
        "complaint_text": complaint_text,
        "recommended_action": predicted_action,
        "responsible_department": responsible_department,
        "action_urgency": action_urgency,
        "confidence_percentage": round(confidence, 2),
        "model_mode": model_mode
    }

    print("\nGovernment Action Recommendation")
    print("Complaint:", complaint_text)
    print("Recommended action:", predicted_action)
    print("Responsible department:", responsible_department)
    print("Action urgency:", action_urgency)
    print("Confidence:", round(confidence, 2), "%")
    print("Model mode:", model_mode)

    return result

sample_complaints = [
    "There is a dangerous pothole on the main road and several vehicles have been damaged.",
    "My bank account was hacked after I shared an OTP with a fake caller.",
    "Garbage has not been collected for one week and sewage is overflowing.",
    "A person is threatening my family and immediate police help is required.",
    "My old age pension application has been pending for three months."
]

sample_recommendations = []

for sample_text in sample_complaints:
    sample_recommendations.append(
        recommend_government_action(sample_text)
    )

sample_recommendations_df = pd.DataFrame(
    sample_recommendations
)

print("\nSample recommendations:")
display(sample_recommendations_df)

sample_recommendations_df.to_csv(
    "10_government_action_sample_predictions.csv",
    index=False
)

sample_recommendations_df.to_sql(
    name="government_action_sample_predictions",
    con=database_engine,
    if_exists="replace",
    index=False
)

mysql_table_names = [
    "government_action_recommendation_results",
    "government_action_summary",
    "government_action_model_evaluation",
    "government_action_label_distribution",
    "government_action_sample_predictions"
]

mysql_table_rows = [
    len(government_action_results_df),
    len(action_summary_df),
    len(evaluation_df),
    len(label_distribution_df),
    len(sample_recommendations_df)
]

if not report_df.empty:
    mysql_table_names.append(
        "government_action_classification_report"
    )
    mysql_table_rows.append(len(report_df))

if not confusion_sql_df.empty:
    mysql_table_names.append(
        "government_action_confusion_matrix"
    )
    mysql_table_rows.append(len(confusion_sql_df))

verification_df = pd.DataFrame({
    "table_name": mysql_table_names,
    "total_rows": mysql_table_rows
})

print("\nMySQL tables created:")
display(verification_df)

sql_sample = pd.read_sql(
    """
    SELECT *
    FROM government_action_recommendation_results
    ORDER BY recommendation_confidence DESC
    LIMIT 10
    """,
    database_engine
)

print("\nGovernment action results read from MySQL:")
display(sql_sample)

print("\n10_Government_Action_Recommendation completed successfully.")
print("\nSource table:", source_table)
print("Selected text columns:", selected_text_columns)
print("Target column:", target_column)
print("Model mode:", model_mode)
print("Total records:", len(government_df))
print(
    "Number of recommended action classes:",
    government_df["predicted_action"].nunique()
)

if evaluation_available:
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("Weighted F1-score:", round(weighted_f1 * 100, 2), "%")

print("\nSaved model:", model_filename)

print("\nCSV output files:")
print("1. 10_government_action_recommendation_results.csv")
print("2. 10_government_action_summary.csv")
print("3. 10_government_action_model_evaluation.csv")
print("4. 10_government_action_label_distribution.csv")
print("5. 10_government_action_sample_predictions.csv")

if not report_df.empty:
    print("6. 10_government_action_classification_report.csv")

if not confusion_sql_df.empty:
    print("7. 10_government_action_confusion_matrix.csv")

print("\nMySQL output tables:")

for index, table_name in enumerate(
    mysql_table_names,
    start=1
):
    print(f"{index}. {table_name}")
```
