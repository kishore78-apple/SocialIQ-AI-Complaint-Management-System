```python
import os
import re
import getpass
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"
database_name = "government_analytics"
source_table = "trend_forecasting"

forecast_periods = 30
minimum_time_points = 15

mysql_password = getpass.getpass(
    "Enter MySQL password: "
)

encoded_password = quote_plus(
    mysql_password
)

server_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)

with server_engine.begin() as connection:
    connection.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS "
            f"`{database_name}`"
        )
    )

database_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/"
    f"{database_name}",
    pool_pre_ping=True
)

with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()

print(
    "Connected database:",
    connected_database
)

existing_tables = inspect(
    database_engine
).get_table_names()

possible_paths = [
    "trend_forecasting.csv",
    "trend_forecasting(1).csv",
    "trend_forecasting(2).csv",
    "08_Trend_Forecasting.csv",
    "government_trends.csv",
    "complaint_trends.csv",
    "citizen_complaints.csv",
    "final_citizen_complaints.csv",
    "/mnt/data/trend_forecasting.csv",
    "/mnt/data/trend_forecasting(1).csv",
    "/mnt/data/trend_forecasting(2).csv",
    "/mnt/data/08_Trend_Forecasting.csv",
    "/mnt/data/government_trends.csv",
    "/mnt/data/complaint_trends.csv",
    "/mnt/data/citizen_complaints.csv",
    "/mnt/data/final_citizen_complaints.csv",
    r"C:\Users\Kishore\Downloads\trend_forecasting.csv",
    r"C:\Users\Kishore\Downloads\trend_forecasting(1).csv",
    r"C:\Users\Kishore\Downloads\trend_forecasting(2).csv",
    r"C:\Users\Kishore\Downloads\08_Trend_Forecasting.csv",
    r"C:\Users\Kishore\Downloads\government_trends.csv",
    r"C:\Users\Kishore\Downloads\complaint_trends.csv",
    r"C:\Users\Kishore\Downloads\citizen_complaints.csv",
    r"C:\Users\Kishore\Downloads\final_citizen_complaints.csv"
]

csv_path = next(
    (
        path
        for path in possible_paths
        if os.path.exists(path)
    ),
    None
)

if csv_path is not None:
    trend_df = pd.read_csv(
        csv_path,
        low_memory=False
    )

    trend_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print(
        "Dataset loaded from CSV:",
        csv_path
    )

    print(
        "Dataset uploaded to MySQL table:",
        source_table
    )

elif source_table in existing_tables:
    trend_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print(
        "Dataset loaded from MySQL table:",
        source_table
    )

else:
    alternative_tables = [
        "citizen_complaints",
        "government_complaints",
        "complaints",
        "unified_training_dataset",
        "government_social_media",
        "social_media_data",
        "harmful_content_detection"
    ]

    selected_table = next(
        (
            table_name
            for table_name in alternative_tables
            if table_name in existing_tables
        ),
        None
    )

    if selected_table is None:
        raise FileNotFoundError(
            "Place trend_forecasting.csv or "
            "08_Trend_Forecasting.csv in the same folder "
            "as your Jupyter Notebook."
        )

    trend_df = pd.read_sql(
        f"SELECT * FROM `{selected_table}`",
        database_engine
    )

    source_table = selected_table

    print(
        "Dataset loaded from alternative MySQL table:",
        source_table
    )

if trend_df.empty:
    raise ValueError(
        "The trend forecasting dataset is empty."
    )

print(
    "\nOriginal dataset shape:",
    trend_df.shape
)

print(
    "\nOriginal columns:"
)

print(
    trend_df.columns.tolist()
)

display(
    trend_df.head()
)

trend_df.columns = (
    trend_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(
        r"[^a-z0-9]+",
        "_",
        regex=True
    )
    .str.strip("_")
)

unnamed_columns = [
    column
    for column in trend_df.columns
    if column.startswith("unnamed")
]

trend_df = trend_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)

print(
    "\nStandardized columns:"
)

print(
    trend_df.columns.tolist()
)

date_candidates = [
    "date",
    "complaint_date",
    "created_date",
    "created_at",
    "submission_date",
    "submitted_date",
    "registration_date",
    "reported_date",
    "post_date",
    "tweet_date",
    "interaction_date",
    "timestamp",
    "datetime",
    "time",
    "record_date",
    "event_date",
    "start_date",
    "application_date",
    "updated_at",
    "year_month",
    "month"
]

date_column = next(
    (
        column
        for column in date_candidates
        if column in trend_df.columns
    ),
    None
)

if date_column is None:
    possible_date_columns = []

    for column in trend_df.columns:
        if any(
            keyword in column
            for keyword in [
                "date",
                "time",
                "month",
                "year"
            ]
        ):
            possible_date_columns.append(column)

    if possible_date_columns:
        date_column = possible_date_columns[0]

if date_column is not None:
    parsed_dates = pd.to_datetime(
        trend_df[date_column],
        errors="coerce",
        dayfirst=True
    )

    valid_date_percentage = (
        parsed_dates.notna().mean()
    )

    if valid_date_percentage >= 0.20:
        trend_df["forecast_date"] = parsed_dates
    else:
        date_column = None

if date_column is None:
    print(
        "\nNo usable date column was found."
    )

    print(
        "Sequential dates are being generated "
        "from the row order."
    )

    start_date = pd.Timestamp(
        "2024-01-01"
    )

    trend_df["forecast_date"] = pd.date_range(
        start=start_date,
        periods=len(trend_df),
        freq="D"
    )

    date_column = "generated_sequential_date"

print(
    "\nSelected date column:",
    date_column
)

numeric_candidates = [
    "complaint_count",
    "record_count",
    "total_complaints",
    "number_of_complaints",
    "complaints",
    "count",
    "frequency",
    "volume",
    "total",
    "trend_value",
    "value",
    "demand",
    "engagement",
    "likes",
    "shares",
    "comments",
    "views",
    "resolution_time_hrs",
    "resolution_time",
    "sentiment_score",
    "severity_score",
    "priority_score"
]

value_column = next(
    (
        column
        for column in numeric_candidates
        if column in trend_df.columns
        and pd.api.types.is_numeric_dtype(
            trend_df[column]
        )
    ),
    None
)

if value_column is None:
    possible_numeric_columns = (
        trend_df.select_dtypes(
            include=[
                "number"
            ]
        )
        .columns
        .tolist()
    )

    possible_numeric_columns = [
        column
        for column in possible_numeric_columns
        if column != "forecast_date"
        and "id" not in column.lower()
        and "code" not in column.lower()
        and trend_df[column].nunique(
            dropna=True
        ) > 1
    ]

    if possible_numeric_columns:
        value_column = possible_numeric_columns[0]

if value_column is not None:
    trend_df[value_column] = pd.to_numeric(
        trend_df[value_column],
        errors="coerce"
    )

    valid_numeric_percentage = (
        trend_df[value_column]
        .notna()
        .mean()
    )

    if valid_numeric_percentage < 0.20:
        value_column = None

print(
    "Selected numeric value column:",
    value_column
)

trend_df = trend_df.dropna(
    subset=[
        "forecast_date"
    ]
).copy()

trend_df["forecast_date"] = pd.to_datetime(
    trend_df["forecast_date"],
    errors="coerce"
)

trend_df = trend_df.dropna(
    subset=[
        "forecast_date"
    ]
).copy()

trend_df["forecast_date"] = (
    trend_df["forecast_date"]
    .dt.tz_localize(None)
    .dt.normalize()
)

if value_column is not None:
    trend_df["trend_value"] = pd.to_numeric(
        trend_df[value_column],
        errors="coerce"
    )

    trend_df["trend_value"] = (
        trend_df["trend_value"]
        .replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )
    )

    median_value = (
        trend_df["trend_value"]
        .median()
    )

    if pd.isna(median_value):
        median_value = 0

    trend_df["trend_value"] = (
        trend_df["trend_value"]
        .fillna(median_value)
    )

    daily_trend = (
        trend_df.groupby(
            "forecast_date",
            as_index=False
        )
        .agg(
            trend_value=(
                "trend_value",
                "sum"
            ),
            record_count=(
                "trend_value",
                "size"
            )
        )
    )

    trend_measure = value_column

else:
    daily_trend = (
        trend_df.groupby(
            "forecast_date"
        )
        .size()
        .reset_index(
            name="trend_value"
        )
    )

    daily_trend["record_count"] = (
        daily_trend["trend_value"]
    )

    trend_measure = "daily_record_count"

daily_trend = daily_trend.sort_values(
    "forecast_date"
).reset_index(
    drop=True
)

if daily_trend.empty:
    raise ValueError(
        "No valid time-series data was available "
        "after date preparation."
    )

full_date_range = pd.date_range(
    start=daily_trend["forecast_date"].min(),
    end=daily_trend["forecast_date"].max(),
    freq="D"
)

daily_trend = (
    daily_trend.set_index(
        "forecast_date"
    )
    .reindex(
        full_date_range
    )
    .rename_axis(
        "forecast_date"
    )
    .reset_index()
)

daily_trend["trend_value"] = (
    daily_trend["trend_value"]
    .fillna(0)
    .astype(float)
)

daily_trend["record_count"] = (
    daily_trend["record_count"]
    .fillna(0)
    .astype(int)
)

if len(daily_trend) < minimum_time_points:
    print(
        "\nThe dataset contains fewer than",
        minimum_time_points,
        "unique dates."
    )

    print(
        "Weekly periods are being expanded into "
        "daily time points for model preparation."
    )

    original_daily_trend = daily_trend.copy()

    repeat_count = int(
        np.ceil(
            minimum_time_points
            / len(original_daily_trend)
        )
    )

    repeated_values = np.tile(
        original_daily_trend[
            "trend_value"
        ].values,
        repeat_count
    )[:minimum_time_points]

    repeated_counts = np.tile(
        original_daily_trend[
            "record_count"
        ].values,
        repeat_count
    )[:minimum_time_points]

    generated_dates = pd.date_range(
        end=original_daily_trend[
            "forecast_date"
        ].max(),
        periods=minimum_time_points,
        freq="D"
    )

    daily_trend = pd.DataFrame({
        "forecast_date":
            generated_dates,
        "trend_value":
            repeated_values,
        "record_count":
            repeated_counts
    })

print(
    "\nTrend measure:",
    trend_measure
)

print(
    "Number of time points:",
    len(daily_trend)
)

print(
    "Start date:",
    daily_trend[
        "forecast_date"
    ].min()
)

print(
    "End date:",
    daily_trend[
        "forecast_date"
    ].max()
)

print(
    "\nPrepared daily trend data:"
)

display(
    daily_trend.head(20)
)

daily_trend["year"] = (
    daily_trend[
        "forecast_date"
    ].dt.year
)

daily_trend["month"] = (
    daily_trend[
        "forecast_date"
    ].dt.month
)

daily_trend["day"] = (
    daily_trend[
        "forecast_date"
    ].dt.day
)

daily_trend["day_of_week"] = (
    daily_trend[
        "forecast_date"
    ].dt.dayofweek
)

daily_trend["day_of_year"] = (
    daily_trend[
        "forecast_date"
    ].dt.dayofyear
)

daily_trend["week_of_year"] = (
    daily_trend[
        "forecast_date"
    ]
    .dt.isocalendar()
    .week
    .astype(int)
)

daily_trend["quarter"] = (
    daily_trend[
        "forecast_date"
    ].dt.quarter
)

daily_trend["is_weekend"] = (
    daily_trend[
        "day_of_week"
    ].isin(
        [
            5,
            6
        ]
    )
    .astype(int)
)

daily_trend["time_index"] = np.arange(
    len(daily_trend)
)

daily_trend["lag_1"] = (
    daily_trend[
        "trend_value"
    ].shift(1)
)

daily_trend["lag_2"] = (
    daily_trend[
        "trend_value"
    ].shift(2)
)

daily_trend["lag_3"] = (
    daily_trend[
        "trend_value"
    ].shift(3)
)

daily_trend["lag_7"] = (
    daily_trend[
        "trend_value"
    ].shift(7)
)

daily_trend["rolling_mean_3"] = (
    daily_trend[
        "trend_value"
    ]
    .shift(1)
    .rolling(
        window=3,
        min_periods=1
    )
    .mean()
)

daily_trend["rolling_mean_7"] = (
    daily_trend[
        "trend_value"
    ]
    .shift(1)
    .rolling(
        window=7,
        min_periods=1
    )
    .mean()
)

daily_trend["rolling_std_7"] = (
    daily_trend[
        "trend_value"
    ]
    .shift(1)
    .rolling(
        window=7,
        min_periods=2
    )
    .std()
)

daily_trend["rolling_min_7"] = (
    daily_trend[
        "trend_value"
    ]
    .shift(1)
    .rolling(
        window=7,
        min_periods=1
    )
    .min()
)

daily_trend["rolling_max_7"] = (
    daily_trend[
        "trend_value"
    ]
    .shift(1)
    .rolling(
        window=7,
        min_periods=1
    )
    .max()
)

daily_trend["trend_change"] = (
    daily_trend[
        "trend_value"
    ].diff()
)

feature_columns = [
    "year",
    "month",
    "day",
    "day_of_week",
    "day_of_year",
    "week_of_year",
    "quarter",
    "is_weekend",
    "time_index",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_7",
    "rolling_mean_3",
    "rolling_mean_7",
    "rolling_std_7",
    "rolling_min_7",
    "rolling_max_7"
]

model_df = daily_trend.dropna(
    subset=feature_columns + [
        "trend_value"
    ]
).reset_index(
    drop=True
)

if len(model_df) < 8:
    reduced_feature_columns = [
        "year",
        "month",
        "day",
        "day_of_week",
        "day_of_year",
        "week_of_year",
        "quarter",
        "is_weekend",
        "time_index",
        "lag_1",
        "lag_2",
        "lag_3",
        "rolling_mean_3"
    ]

    model_df = daily_trend.dropna(
        subset=reduced_feature_columns + [
            "trend_value"
        ]
    ).reset_index(
        drop=True
    )

    feature_columns = reduced_feature_columns

if len(model_df) < 5:
    raise ValueError(
        "The dataset does not contain enough time points "
        "for trend forecasting. At least five prepared "
        "time points are required."
    )

print(
    "\nModel-ready records:",
    len(model_df)
)

print(
    "\nForecasting features:"
)

print(
    feature_columns
)

test_size = max(
    2,
    int(
        round(
            len(model_df) * 0.20
        )
    )
)

if test_size >= len(model_df):
    test_size = max(
        1,
        len(model_df) // 3
    )

train_df = model_df.iloc[
    :-test_size
].copy()

test_df = model_df.iloc[
    -test_size:
].copy()

if train_df.empty:
    raise ValueError(
        "The training dataset is empty after "
        "time-based splitting."
    )

X_train = train_df[
    feature_columns
]

y_train = train_df[
    "trend_value"
]

X_test = test_df[
    feature_columns
]

y_test = test_df[
    "trend_value"
]

print(
    "\nTraining records:",
    len(X_train)
)

print(
    "Testing records:",
    len(X_test)
)

trend_forecasting_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

trend_forecasting_model.fit(
    X_train,
    y_train
)

print(
    "\nTrend Forecasting model "
    "trained successfully."
)

y_pred = trend_forecasting_model.predict(
    X_test
)

y_pred = np.maximum(
    y_pred,
    0
)

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mse
)

if len(y_test) >= 2:
    r2 = r2_score(
        y_test,
        y_pred
    )
else:
    r2 = np.nan

non_zero_mask = (
    y_test.values != 0
)

if non_zero_mask.any():
    mape = np.mean(
        np.abs(
            (
                y_test.values[
                    non_zero_mask
                ]
                -
                y_pred[
                    non_zero_mask
                ]
            )
            /
            y_test.values[
                non_zero_mask
            ]
        )
    ) * 100
else:
    mape = np.nan

print(
    "\nTrend Forecasting Model Evaluation"
)

print(
    "Mean Absolute Error:",
    round(
        mae,
        4
    )
)

print(
    "Mean Squared Error:",
    round(
        mse,
        4
    )
)

print(
    "Root Mean Squared Error:",
    round(
        rmse,
        4
    )
)

if pd.notna(r2):
    print(
        "R-squared:",
        round(
            r2,
            4
        )
    )
else:
    print(
        "R-squared: Not available"
    )

if pd.notna(mape):
    print(
        "Mean Absolute Percentage Error:",
        round(
            mape,
            2
        ),
        "%"
    )
else:
    print(
        "Mean Absolute Percentage Error: "
        "Not available because actual values are zero."
    )

test_results = pd.DataFrame({
    "forecast_date":
        test_df[
            "forecast_date"
        ].values,
    "actual_value":
        y_test.values,
    "predicted_value":
        y_pred
})

test_results["absolute_error"] = np.abs(
    test_results[
        "actual_value"
    ]
    -
    test_results[
        "predicted_value"
    ]
)

test_results["squared_error"] = (
    test_results[
        "absolute_error"
    ] ** 2
)

test_results["prediction_difference"] = (
    test_results[
        "predicted_value"
    ]
    -
    test_results[
        "actual_value"
    ]
)

test_results["predicted_value"] = (
    test_results[
        "predicted_value"
    ].round(2)
)

test_results["absolute_error"] = (
    test_results[
        "absolute_error"
    ].round(2)
)

test_results["squared_error"] = (
    test_results[
        "squared_error"
    ].round(2)
)

test_results[
    "prediction_difference"
] = (
    test_results[
        "prediction_difference"
    ].round(2)
)

test_results.insert(
    0,
    "prediction_id",
    range(
        1,
        len(test_results) + 1
    )
)

print(
    "\nTest prediction results:"
)

display(
    test_results
)

plt.figure(
    figsize=(
        14,
        6
    )
)

plt.plot(
    test_results[
        "forecast_date"
    ],
    test_results[
        "actual_value"
    ],
    marker="o",
    label="Actual"
)

plt.plot(
    test_results[
        "forecast_date"
    ],
    test_results[
        "predicted_value"
    ],
    marker="o",
    label="Predicted"
)

plt.title(
    "Actual vs Predicted Trend Values"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Trend Value"
)

plt.xticks(
    rotation=45
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()

forecast_history = daily_trend[
    [
        "forecast_date",
        "trend_value"
    ]
].copy()

future_forecasts = []

last_date = forecast_history[
    "forecast_date"
].max()

historical_values = (
    forecast_history[
        "trend_value"
    ]
    .astype(float)
    .tolist()
)

for step in range(
    1,
    forecast_periods + 1
):
    future_date = (
        last_date
        +
        pd.Timedelta(
            days=step
        )
    )

    lag_1 = (
        historical_values[-1]
        if len(historical_values) >= 1
        else 0
    )

    lag_2 = (
        historical_values[-2]
        if len(historical_values) >= 2
        else lag_1
    )

    lag_3 = (
        historical_values[-3]
        if len(historical_values) >= 3
        else lag_2
    )

    lag_7 = (
        historical_values[-7]
        if len(historical_values) >= 7
        else historical_values[0]
    )

    recent_3 = historical_values[
        -3:
    ]

    recent_7 = historical_values[
        -7:
    ]

    rolling_mean_3 = float(
        np.mean(
            recent_3
        )
    )

    rolling_mean_7 = float(
        np.mean(
            recent_7
        )
    )

    rolling_std_7 = float(
        np.std(
            recent_7,
            ddof=1
        )
    ) if len(recent_7) >= 2 else 0.0

    rolling_min_7 = float(
        np.min(
            recent_7
        )
    )

    rolling_max_7 = float(
        np.max(
            recent_7
        )
    )

    feature_record = {
        "year":
            future_date.year,
        "month":
            future_date.month,
        "day":
            future_date.day,
        "day_of_week":
            future_date.dayofweek,
        "day_of_year":
            future_date.dayofyear,
        "week_of_year":
            int(
                future_date.isocalendar().week
            ),
        "quarter":
            future_date.quarter,
        "is_weekend":
            int(
                future_date.dayofweek
                in [
                    5,
                    6
                ]
            ),
        "time_index":
            len(daily_trend)
            +
            step
            -
            1,
        "lag_1":
            lag_1,
        "lag_2":
            lag_2,
        "lag_3":
            lag_3,
        "lag_7":
            lag_7,
        "rolling_mean_3":
            rolling_mean_3,
        "rolling_mean_7":
            rolling_mean_7,
        "rolling_std_7":
            rolling_std_7,
        "rolling_min_7":
            rolling_min_7,
        "rolling_max_7":
            rolling_max_7
    }

    future_feature_df = pd.DataFrame(
        [
            {
                column:
                    feature_record.get(
                        column,
                        0
                    )
                for column in feature_columns
            }
        ]
    )

    predicted_future_value = float(
        trend_forecasting_model.predict(
            future_feature_df
        )[0]
    )

    predicted_future_value = max(
        0,
        predicted_future_value
    )

    historical_values.append(
        predicted_future_value
    )

    future_forecasts.append({
        "forecast_date":
            future_date,
        "forecasted_value":
            round(
                predicted_future_value,
                2
            )
    })

future_forecast_df = pd.DataFrame(
    future_forecasts
)

future_forecast_df.insert(
    0,
    "forecast_id",
    range(
        1,
        len(future_forecast_df) + 1
    )
)

future_forecast_df[
    "forecast_year"
] = (
    future_forecast_df[
        "forecast_date"
    ].dt.year
)

future_forecast_df[
    "forecast_month"
] = (
    future_forecast_df[
        "forecast_date"
    ].dt.month
)

future_forecast_df[
    "forecast_day"
] = (
    future_forecast_df[
        "forecast_date"
    ].dt.day
)

future_forecast_df[
    "day_name"
] = (
    future_forecast_df[
        "forecast_date"
    ].dt.day_name()
)

average_historical_value = (
    daily_trend[
        "trend_value"
    ].mean()
)

future_forecast_df[
    "trend_direction"
] = np.where(
    future_forecast_df[
        "forecasted_value"
    ]
    >
    average_historical_value,
    "Above Historical Average",
    np.where(
        future_forecast_df[
            "forecasted_value"
        ]
        <
        average_historical_value,
        "Below Historical Average",
        "Near Historical Average"
    )
)

print(
    "\nFuture trend forecast:"
)

display(
    future_forecast_df
)

plt.figure(
    figsize=(
        15,
        7
    )
)

historical_plot_data = daily_trend.tail(
    min(
        90,
        len(daily_trend)
    )
)

plt.plot(
    historical_plot_data[
        "forecast_date"
    ],
    historical_plot_data[
        "trend_value"
    ],
    marker="o",
    label="Historical Trend"
)

plt.plot(
    future_forecast_df[
        "forecast_date"
    ],
    future_forecast_df[
        "forecasted_value"
    ],
    marker="o",
    label="Future Forecast"
)

plt.axvline(
    x=daily_trend[
        "forecast_date"
    ].max(),
    linestyle="--",
    label="Forecast Start"
)

plt.title(
    "08 Trend Forecasting"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Trend Value"
)

plt.xticks(
    rotation=45
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()

feature_importance_df = pd.DataFrame({
    "feature":
        feature_columns,
    "importance":
        trend_forecasting_model
        .feature_importances_
})

feature_importance_df = (
    feature_importance_df
    .sort_values(
        by="importance",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

feature_importance_df[
    "importance_percentage"
] = (
    feature_importance_df[
        "importance"
    ] * 100
).round(2)

print(
    "\nFeature importance:"
)

display(
    feature_importance_df
)

plt.figure(
    figsize=(
        10,
        7
    )
)

plt.barh(
    feature_importance_df[
        "feature"
    ][::-1],
    feature_importance_df[
        "importance_percentage"
    ][::-1]
)

plt.title(
    "Trend Forecasting Feature Importance"
)

plt.xlabel(
    "Importance Percentage"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()

plt.show()

monthly_summary_df = (
    daily_trend.assign(
        year_month=(
            daily_trend[
                "forecast_date"
            ]
            .dt.to_period(
                "M"
            )
            .astype(str)
        )
    )
    .groupby(
        "year_month",
        as_index=False
    )
    .agg(
        total_trend_value=(
            "trend_value",
            "sum"
        ),
        average_trend_value=(
            "trend_value",
            "mean"
        ),
        maximum_trend_value=(
            "trend_value",
            "max"
        ),
        minimum_trend_value=(
            "trend_value",
            "min"
        ),
        number_of_days=(
            "trend_value",
            "size"
        )
    )
)

monthly_summary_df[
    "total_trend_value"
] = (
    monthly_summary_df[
        "total_trend_value"
    ].round(2)
)

monthly_summary_df[
    "average_trend_value"
] = (
    monthly_summary_df[
        "average_trend_value"
    ].round(2)
)

monthly_summary_df[
    "maximum_trend_value"
] = (
    monthly_summary_df[
        "maximum_trend_value"
    ].round(2)
)

monthly_summary_df[
    "minimum_trend_value"
] = (
    monthly_summary_df[
        "minimum_trend_value"
    ].round(2)
)

print(
    "\nMonthly trend summary:"
)

display(
    monthly_summary_df
)

evaluation_df = pd.DataFrame({
    "metric": [
        "mean_absolute_error",
        "mean_squared_error",
        "root_mean_squared_error",
        "r_squared",
        "mean_absolute_percentage_error",
        "original_records",
        "daily_time_points",
        "model_ready_records",
        "training_records",
        "testing_records",
        "forecast_periods",
        "selected_date_column",
        "selected_value_column",
        "trend_measure"
    ],
    "value": [
        str(
            float(mae)
        ),
        str(
            float(mse)
        ),
        str(
            float(rmse)
        ),
        (
            str(
                float(r2)
            )
            if pd.notna(r2)
            else "Not Available"
        ),
        (
            str(
                float(mape)
            )
            if pd.notna(mape)
            else "Not Available"
        ),
        str(
            int(
                len(trend_df)
            )
        ),
        str(
            int(
                len(daily_trend)
            )
        ),
        str(
            int(
                len(model_df)
            )
        ),
        str(
            int(
                len(X_train)
            )
        ),
        str(
            int(
                len(X_test)
            )
        ),
        str(
            int(
                forecast_periods
            )
        ),
        str(
            date_column
        ),
        str(
            value_column
            if value_column is not None
            else "Daily Record Count"
        ),
        str(
            trend_measure
        )
    ]
})

print(
    "\nModel evaluation summary:"
)

display(
    evaluation_df
)

prepared_trend_data = daily_trend.copy()

prepared_trend_data.insert(
    0,
    "trend_record_id",
    range(
        1,
        len(
            prepared_trend_data
        ) + 1
    )
)

model_package = {
    "model":
        trend_forecasting_model,
    "feature_columns":
        feature_columns,
    "date_column":
        date_column,
    "value_column":
        value_column,
    "trend_measure":
        trend_measure,
    "forecast_periods":
        forecast_periods,
    "last_date":
        daily_trend[
            "forecast_date"
        ].max(),
    "historical_values":
        daily_trend[
            "trend_value"
        ].tolist()
}

model_filename = (
    "08_trend_forecasting_model.pkl"
)

joblib.dump(
    model_package,
    model_filename
)

print(
    "\nModel saved successfully:",
    model_filename
)

test_results.to_csv(
    "08_trend_forecasting_test_results.csv",
    index=False
)

future_forecast_df.to_csv(
    "08_trend_forecasting_future_predictions.csv",
    index=False
)

evaluation_df.to_csv(
    "08_trend_forecasting_model_evaluation.csv",
    index=False
)

prepared_trend_data.to_csv(
    "08_trend_forecasting_prepared_data.csv",
    index=False
)

monthly_summary_df.to_csv(
    "08_trend_forecasting_monthly_summary.csv",
    index=False
)

feature_importance_df.to_csv(
    "08_trend_forecasting_feature_importance.csv",
    index=False
)

print(
    "\nCSV files saved successfully."
)

test_results.to_sql(
    name="trend_forecasting_test_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

future_forecast_df.to_sql(
    name="trend_forecasting_future_predictions",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

evaluation_df.to_sql(
    name="trend_forecasting_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)

prepared_trend_data.to_sql(
    name="trend_forecasting_prepared_data",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

monthly_summary_df.to_sql(
    name="trend_forecasting_monthly_summary",
    con=database_engine,
    if_exists="replace",
    index=False
)

feature_importance_df.to_sql(
    name="trend_forecasting_feature_importance",
    con=database_engine,
    if_exists="replace",
    index=False
)

print(
    "\nAll Trend Forecasting tables "
    "created in MySQL."
)

def forecast_next_days(number_of_days=7):
    if not isinstance(
        number_of_days,
        int
    ):
        raise TypeError(
            "Number of days must be an integer."
        )

    if number_of_days < 1:
        raise ValueError(
            "Number of days must be at least 1."
        )

    if number_of_days > 365:
        raise ValueError(
            "A maximum of 365 days can be forecast."
        )

    temporary_history = (
        daily_trend[
            "trend_value"
        ]
        .astype(float)
        .tolist()
    )

    last_historical_date = (
        daily_trend[
            "forecast_date"
        ].max()
    )

    predictions = []

    for step in range(
        1,
        number_of_days + 1
    ):
        prediction_date = (
            last_historical_date
            +
            pd.Timedelta(
                days=step
            )
        )

        lag_1 = (
            temporary_history[-1]
            if len(temporary_history) >= 1
            else 0
        )

        lag_2 = (
            temporary_history[-2]
            if len(temporary_history) >= 2
            else lag_1
        )

        lag_3 = (
            temporary_history[-3]
            if len(temporary_history) >= 3
            else lag_2
        )

        lag_7 = (
            temporary_history[-7]
            if len(temporary_history) >= 7
            else temporary_history[0]
        )

        recent_3 = temporary_history[
            -3:
        ]

        recent_7 = temporary_history[
            -7:
        ]

        feature_record = {
            "year":
                prediction_date.year,
            "month":
                prediction_date.month,
            "day":
                prediction_date.day,
            "day_of_week":
                prediction_date.dayofweek,
            "day_of_year":
                prediction_date.dayofyear,
            "week_of_year":
                int(
                    prediction_date
                    .isocalendar()
                    .week
                ),
            "quarter":
                prediction_date.quarter,
            "is_weekend":
                int(
                    prediction_date.dayofweek
                    in [
                        5,
                        6
                    ]
                ),
            "time_index":
                len(daily_trend)
                +
                step
                -
                1,
            "lag_1":
                lag_1,
            "lag_2":
                lag_2,
            "lag_3":
                lag_3,
            "lag_7":
                lag_7,
            "rolling_mean_3":
                float(
                    np.mean(
                        recent_3
                    )
                ),
            "rolling_mean_7":
                float(
                    np.mean(
                        recent_7
                    )
                ),
            "rolling_std_7":
                (
                    float(
                        np.std(
                            recent_7,
                            ddof=1
                        )
                    )
                    if len(recent_7) >= 2
                    else 0.0
                ),
            "rolling_min_7":
                float(
                    np.min(
                        recent_7
                    )
                ),
            "rolling_max_7":
                float(
                    np.max(
                        recent_7
                    )
                )
        }

        prediction_features = pd.DataFrame(
            [
                {
                    column:
                        feature_record.get(
                            column,
                            0
                        )
                    for column in feature_columns
                }
            ]
        )

        predicted_value = float(
            trend_forecasting_model.predict(
                prediction_features
            )[0]
        )

        predicted_value = max(
            0,
            predicted_value
        )

        temporary_history.append(
            predicted_value
        )

        predictions.append({
            "forecast_date":
                prediction_date,
            "forecasted_value":
                round(
                    predicted_value,
                    2
                )
        })

    prediction_df = pd.DataFrame(
        predictions
    )

    prediction_df.insert(
        0,
        "forecast_id",
        range(
            1,
            len(prediction_df) + 1
        )
    )

    print(
        f"\nForecast for the next "
        f"{number_of_days} days:"
    )

    display(
        prediction_df
    )

    return prediction_df

forecast_next_days(
    7
)

verification_df = pd.DataFrame({
    "table_name": [
        "trend_forecasting_test_results",
        "trend_forecasting_future_predictions",
        "trend_forecasting_model_evaluation",
        "trend_forecasting_prepared_data",
        "trend_forecasting_monthly_summary",
        "trend_forecasting_feature_importance"
    ],
    "total_rows": [
        len(
            test_results
        ),
        len(
            future_forecast_df
        ),
        len(
            evaluation_df
        ),
        len(
            prepared_trend_data
        ),
        len(
            monthly_summary_df
        ),
        len(
            feature_importance_df
        )
    ]
})

print(
    "\nMySQL tables created:"
)

display(
    verification_df
)

sql_forecast_sample = pd.read_sql(
    """
    SELECT *
    FROM trend_forecasting_future_predictions
    LIMIT 10
    """,
    database_engine
)

print(
    "\nSample future forecasts "
    "read from MySQL:"
)

display(
    sql_forecast_sample
)

print(
    "\n08_Trend_Forecasting "
    "completed successfully."
)

print(
    "\nSource table:",
    source_table
)

print(
    "Selected date column:",
    date_column
)

print(
    "Selected value column:",
    (
        value_column
        if value_column is not None
        else "Daily Record Count"
    )
)

print(
    "Trend measure:",
    trend_measure
)

print(
    "Original records:",
    len(trend_df)
)

print(
    "Daily time points:",
    len(daily_trend)
)

print(
    "Model-ready records:",
    len(model_df)
)

print(
    "Training records:",
    len(X_train)
)

print(
    "Testing records:",
    len(X_test)
)

print(
    "Forecast periods:",
    forecast_periods
)

print(
    "Mean Absolute Error:",
    round(
        mae,
        4
    )
)

print(
    "Root Mean Squared Error:",
    round(
        rmse,
        4
    )
)

if pd.notna(r2):
    print(
        "R-squared:",
        round(
            r2,
            4
        )
    )

print(
    "\nSaved model:",
    model_filename
)

print(
    "\nCSV output files:"
)

print(
    "1. 08_trend_forecasting_test_results.csv"
)

print(
    "2. 08_trend_forecasting_future_predictions.csv"
)

print(
    "3. 08_trend_forecasting_model_evaluation.csv"
)

print(
    "4. 08_trend_forecasting_prepared_data.csv"
)

print(
    "5. 08_trend_forecasting_monthly_summary.csv"
)

print(
    "6. 08_trend_forecasting_feature_importance.csv"
)

print(
    "\nMySQL output tables:"
)

print(
    "1. trend_forecasting_test_results"
)

print(
    "2. trend_forecasting_future_predictions"
)

print(
    "3. trend_forecasting_model_evaluation"
)

print(
    "4. trend_forecasting_prepared_data"
)

print(
    "5. trend_forecasting_monthly_summary"
)

print(
    "6. trend_forecasting_feature_importance"
)
```
