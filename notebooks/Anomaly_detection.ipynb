```python
import os
import re
import getpass
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"

database_name = "government_analytics"
source_table = "anomaly_detection"

contamination_rate = 0.05
minimum_records_required = 10

mysql_password = getpass.getpass(
    "Enter MySQL password: "
)

encoded_password = quote_plus(
    mysql_password
)

server_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)

with server_engine.begin() as connection:
    connection.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS "
            f"`{database_name}`"
        )
    )

database_engine = create_engine(
    f"mysql+mysqlconnector://"
    f"{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/"
    f"{database_name}",
    pool_pre_ping=True
)

with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()

print(
    "Connected database:",
    connected_database
)

existing_tables = inspect(
    database_engine
).get_table_names()

possible_paths = [
    "anomaly_detection.csv",
    "anomaly_detection(1).csv",
    "anomaly_detection(2).csv",
    "09_Anomaly_Detection.csv",
    "government_anomaly_data.csv",
    "government_analytics_data.csv",
    "citizen_complaints.csv",
    "final_citizen_complaints.csv",
    "trend_forecasting.csv",

    "/mnt/data/anomaly_detection.csv",
    "/mnt/data/anomaly_detection(1).csv",
    "/mnt/data/anomaly_detection(2).csv",
    "/mnt/data/09_Anomaly_Detection.csv",
    "/mnt/data/government_anomaly_data.csv",
    "/mnt/data/government_analytics_data.csv",
    "/mnt/data/citizen_complaints.csv",
    "/mnt/data/final_citizen_complaints.csv",
    "/mnt/data/trend_forecasting.csv",

    r"C:\Users\Kishore\Downloads\anomaly_detection.csv",
    r"C:\Users\Kishore\Downloads\anomaly_detection(1).csv",
    r"C:\Users\Kishore\Downloads\anomaly_detection(2).csv",
    r"C:\Users\Kishore\Downloads\09_Anomaly_Detection.csv",
    r"C:\Users\Kishore\Downloads\government_anomaly_data.csv",
    r"C:\Users\Kishore\Downloads\citizen_complaints.csv",
    r"C:\Users\Kishore\Downloads\final_citizen_complaints.csv"
]

csv_path = next(
    (
        path
        for path in possible_paths
        if os.path.exists(path)
    ),
    None
)

if csv_path is not None:
    anomaly_df = pd.read_csv(
        csv_path,
        low_memory=False
    )

    anomaly_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print(
        "Dataset loaded from CSV:",
        csv_path
    )

    print(
        "Dataset uploaded to MySQL table:",
        source_table
    )

elif source_table in existing_tables:
    anomaly_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print(
        "Dataset loaded from MySQL table:",
        source_table
    )

else:
    alternative_tables = [
        "trend_forecasting",
        "citizen_complaints",
        "government_complaints",
        "complaints",
        "unified_training_dataset",
        "government_social_media",
        "harmful_content_detection",
        "emergency_detection",
        "priority_prediction"
    ]

    selected_table = next(
        (
            table_name
            for table_name in alternative_tables
            if table_name in existing_tables
        ),
        None
    )

    if selected_table is None:
        raise FileNotFoundError(
            "Place anomaly_detection.csv or "
            "09_Anomaly_Detection.csv in the same folder "
            "as the Jupyter Notebook."
        )

    anomaly_df = pd.read_sql(
        f"SELECT * FROM `{selected_table}`",
        database_engine
    )

    source_table = selected_table

    print(
        "Dataset loaded from alternative MySQL table:",
        source_table
    )

if anomaly_df.empty:
    raise ValueError(
        "The anomaly detection dataset is empty."
    )

print(
    "\nOriginal dataset shape:",
    anomaly_df.shape
)

print(
    "\nOriginal columns:"
)

print(
    anomaly_df.columns.tolist()
)

display(
    anomaly_df.head()
)

anomaly_df.columns = (
    anomaly_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(
        r"[^a-z0-9]+",
        "_",
        regex=True
    )
    .str.strip("_")
)

unnamed_columns = [
    column
    for column in anomaly_df.columns
    if column.startswith("unnamed")
]

anomaly_df = anomaly_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)

print(
    "\nStandardized columns:"
)

print(
    anomaly_df.columns.tolist()
)

date_candidates = [
    "date",
    "complaint_date",
    "created_date",
    "created_at",
    "submission_date",
    "submitted_date",
    "registration_date",
    "reported_date",
    "post_date",
    "tweet_date",
    "interaction_date",
    "timestamp",
    "datetime",
    "record_date",
    "event_date",
    "start_date",
    "updated_at"
]

date_column = next(
    (
        column
        for column in date_candidates
        if column in anomaly_df.columns
    ),
    None
)

if date_column is None:
    possible_date_columns = [
        column
        for column in anomaly_df.columns
        if any(
            word in column
            for word in [
                "date",
                "time",
                "created",
                "reported"
            ]
        )
    ]

    if possible_date_columns:
        date_column = possible_date_columns[0]

if date_column is not None:
    parsed_dates = pd.to_datetime(
        anomaly_df[date_column],
        errors="coerce",
        dayfirst=True
    )

    valid_date_ratio = (
        parsed_dates.notna().mean()
    )

    if valid_date_ratio >= 0.20:
        anomaly_df["anomaly_date"] = parsed_dates
    else:
        date_column = None

if date_column is None:
    anomaly_df["anomaly_date"] = pd.date_range(
        start="2024-01-01",
        periods=len(anomaly_df),
        freq="D"
    )

    date_column = "generated_sequential_date"

print(
    "\nSelected date column:",
    date_column
)

for column in anomaly_df.columns:
    if column in [
        "anomaly_date"
    ]:
        continue

    if anomaly_df[column].dtype == "object":
        cleaned_numeric = (
            anomaly_df[column]
            .astype(str)
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.replace(
                "%",
                "",
                regex=False
            )
            .str.replace(
                r"[^\d\.\-]",
                "",
                regex=True
            )
        )

        converted_values = pd.to_numeric(
            cleaned_numeric,
            errors="coerce"
        )

        valid_numeric_ratio = (
            converted_values.notna().mean()
        )

        if valid_numeric_ratio >= 0.80:
            anomaly_df[column] = converted_values

numeric_columns = (
    anomaly_df.select_dtypes(
        include=[
            "number"
        ]
    )
    .columns
    .tolist()
)

excluded_keywords = [
    "id",
    "code",
    "year",
    "month",
    "day",
    "index",
    "serial",
    "phone",
    "mobile",
    "pincode",
    "zip"
]

feature_columns = []

for column in numeric_columns:
    if column == "anomaly_date":
        continue

    if any(
        keyword in column.lower()
        for keyword in excluded_keywords
    ):
        continue

    unique_values = anomaly_df[
        column
    ].nunique(
        dropna=True
    )

    if unique_values <= 1:
        continue

    valid_ratio = anomaly_df[
        column
    ].notna().mean()

    if valid_ratio < 0.20:
        continue

    feature_columns.append(
        column
    )

preferred_features = [
    "complaint_count",
    "record_count",
    "resolution_time_hrs",
    "resolution_time",
    "sentiment_score",
    "severity_score",
    "priority_score",
    "likes",
    "shares",
    "comments",
    "views",
    "engagement",
    "response_time",
    "processing_time",
    "amount",
    "transaction_amount",
    "frequency",
    "volume",
    "total"
]

preferred_available = [
    column
    for column in preferred_features
    if column in feature_columns
]

remaining_features = [
    column
    for column in feature_columns
    if column not in preferred_available
]

feature_columns = (
    preferred_available
    +
    remaining_features
)

maximum_features = 15

feature_columns = feature_columns[
    :maximum_features
]

if not feature_columns:
    raise ValueError(
        "No usable numeric feature columns were found. "
        "The dataset must contain numeric values such as "
        "complaint count, resolution time, score, amount, "
        "frequency, views, likes or engagement."
    )

print(
    "\nSelected anomaly detection features:"
)

print(
    feature_columns
)

model_df = anomaly_df.copy()

for column in feature_columns:
    model_df[column] = pd.to_numeric(
        model_df[column],
        errors="coerce"
    )

    model_df[column] = model_df[
        column
    ].replace(
        [
            np.inf,
            -np.inf
        ],
        np.nan
    )

    median_value = model_df[
        column
    ].median()

    if pd.isna(median_value):
        median_value = 0

    model_df[column] = model_df[
        column
    ].fillna(
        median_value
    )

prepared_model_df = model_df.reset_index(
    drop=True
).copy()

deduplicated_model_df = model_df.drop_duplicates(
    subset=feature_columns
).reset_index(
    drop=True
)

if len(deduplicated_model_df) >= minimum_records_required:
    model_df = deduplicated_model_df

else:
    print(
        "\nOnly",
        len(deduplicated_model_df),
        "unique numeric records were found."
    )

    print(
        "Duplicate rows will be retained so that "
        "the anomaly detection model can continue."
    )

    model_df = prepared_model_df

if len(model_df) < 2:
    raise ValueError(
        "The dataset must contain at least two usable records "
        "after numeric data preparation."
    )

print(
    "\nOriginal records:",
    len(anomaly_df)
)

print(
    "Usable records:",
    len(model_df)
)

print(
    "Number of selected features:",
    len(feature_columns)
)

print(
    "\nPrepared numeric data:"
)

display(
    model_df[
        feature_columns
    ].head(10)
)

label_candidates = [
    "anomaly_label",
    "anomaly_status",
    "is_anomaly",
    "anomaly",
    "fraud_label",
    "outlier_label",
    "outlier_status",
    "is_outlier",
    "target",
    "label",
    "class"
]

target_column = next(
    (
        column
        for column in label_candidates
        if column in model_df.columns
    ),
    None
)

if target_column is not None:
    print(
        "\nExisting target column detected:",
        target_column
    )

    display(
        model_df[
            target_column
        ]
        .value_counts(
            dropna=False
        )
        .rename_axis(
            "original_label"
        )
        .reset_index(
            name="record_count"
        )
    )

else:
    print(
        "\nNo existing anomaly label was found."
    )

    print(
        "The model will perform unsupervised "
        "anomaly detection."
    )

X = model_df[
    feature_columns
].copy()

anomaly_detection_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "isolation_forest",
        IsolationForest(
            n_estimators=300,
            contamination=min(
                contamination_rate,
                max(
                    1 / len(model_df),
                    0.001
                )
            ),
            max_samples="auto",
            random_state=42,
            n_jobs=-1
        )
    )
])

anomaly_detection_model.fit(
    X
)

print(
    "\nIsolation Forest anomaly detection "
    "model trained successfully."
)

raw_predictions = anomaly_detection_model.predict(
    X
)

decision_scores = anomaly_detection_model.decision_function(
    X
)

model_df["anomaly_prediction_code"] = (
    raw_predictions
)

model_df["anomaly_status"] = np.where(
    raw_predictions == -1,
    "Anomaly",
    "Normal"
)

model_df["anomaly_score"] = (
    -decision_scores
)

minimum_score = model_df[
    "anomaly_score"
].min()

maximum_score = model_df[
    "anomaly_score"
].max()

if maximum_score > minimum_score:
    model_df[
        "anomaly_score_percentage"
    ] = (
        (
            model_df[
                "anomaly_score"
            ]
            -
            minimum_score
        )
        /
        (
            maximum_score
            -
            minimum_score
        )
        *
        100
    )
else:
    model_df[
        "anomaly_score_percentage"
    ] = 0.0

model_df[
    "anomaly_score_percentage"
] = (
    model_df[
        "anomaly_score_percentage"
    ].round(2)
)

model_df["anomaly_rank"] = (
    model_df[
        "anomaly_score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)

model_df["risk_level"] = pd.cut(
    model_df[
        "anomaly_score_percentage"
    ],
    bins=[
        -np.inf,
        25,
        50,
        75,
        np.inf
    ],
    labels=[
        "Low",
        "Medium",
        "High",
        "Critical"
    ]
).astype(str)

model_df.insert(
    0,
    "anomaly_record_id",
    range(
        1,
        len(model_df) + 1
    )
)

distribution_df = (
    model_df[
        "anomaly_status"
    ]
    .value_counts()
    .rename_axis(
        "anomaly_status"
    )
    .reset_index(
        name="record_count"
    )
)

distribution_df[
    "percentage"
] = (
    distribution_df[
        "record_count"
    ]
    /
    len(model_df)
    *
    100
).round(2)

print(
    "\nAnomaly distribution:"
)

display(
    distribution_df
)

normal_count = int(
    (
        model_df[
            "anomaly_status"
        ]
        ==
        "Normal"
    ).sum()
)

anomaly_count = int(
    (
        model_df[
            "anomaly_status"
        ]
        ==
        "Anomaly"
    ).sum()
)

anomaly_percentage = (
    anomaly_count
    /
    len(model_df)
    *
    100
)

print(
    "\nNormal records:",
    normal_count
)

print(
    "Anomaly records:",
    anomaly_count
)

print(
    "Detected anomaly percentage:",
    round(
        anomaly_percentage,
        2
    ),
    "%"
)

top_anomalies_df = (
    model_df[
        model_df[
            "anomaly_status"
        ]
        ==
        "Anomaly"
    ]
    .sort_values(
        by="anomaly_score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print(
    "\nTop detected anomalies:"
)

display_columns = [
    "anomaly_record_id",
    "anomaly_status",
    "anomaly_score",
    "anomaly_score_percentage",
    "anomaly_rank",
    "risk_level"
] + feature_columns

display(
    top_anomalies_df[
        display_columns
    ].head(20)
)

feature_summary_df = (
    model_df[
        feature_columns
    ]
    .describe()
    .transpose()
    .reset_index()
    .rename(
        columns={
            "index": "feature"
        }
    )
)

feature_summary_df[
    "missing_values"
] = [
    int(
        anomaly_df[
            column
        ].isna().sum()
    )
    for column in feature_columns
]

feature_summary_df[
    "unique_values"
] = [
    int(
        anomaly_df[
            column
        ].nunique(
            dropna=True
        )
    )
    for column in feature_columns
]

print(
    "\nFeature statistical summary:"
)

display(
    feature_summary_df
)

scaled_values = anomaly_detection_model.named_steps[
    "scaler"
].transform(
    X
)

absolute_z_scores = np.abs(
    scaled_values
)

average_feature_deviation = (
    absolute_z_scores.mean(
        axis=0
    )
)

feature_contribution_df = pd.DataFrame({
    "feature":
        feature_columns,
    "average_absolute_standardized_deviation":
        average_feature_deviation
})

feature_contribution_df = (
    feature_contribution_df
    .sort_values(
        by="average_absolute_standardized_deviation",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

total_contribution = feature_contribution_df[
    "average_absolute_standardized_deviation"
].sum()

if total_contribution > 0:
    feature_contribution_df[
        "contribution_percentage"
    ] = (
        feature_contribution_df[
            "average_absolute_standardized_deviation"
        ]
        /
        total_contribution
        *
        100
    ).round(2)
else:
    feature_contribution_df[
        "contribution_percentage"
    ] = 0.0

print(
    "\nFeature anomaly contribution:"
)

display(
    feature_contribution_df
)

plt.figure(
    figsize=(
        8,
        5
    )
)

plt.bar(
    distribution_df[
        "anomaly_status"
    ],
    distribution_df[
        "record_count"
    ]
)

plt.title(
    "09 Anomaly Detection Distribution"
)

plt.xlabel(
    "Anomaly Status"
)

plt.ylabel(
    "Number of Records"
)

plt.tight_layout()

plt.show()

plot_df = model_df.sort_values(
    by="anomaly_record_id"
)

plt.figure(
    figsize=(
        14,
        6
    )
)

normal_records = plot_df[
    plot_df[
        "anomaly_status"
    ]
    ==
    "Normal"
]

anomaly_records = plot_df[
    plot_df[
        "anomaly_status"
    ]
    ==
    "Anomaly"
]

plt.scatter(
    normal_records[
        "anomaly_record_id"
    ],
    normal_records[
        "anomaly_score_percentage"
    ],
    label="Normal",
    alpha=0.7
)

plt.scatter(
    anomaly_records[
        "anomaly_record_id"
    ],
    anomaly_records[
        "anomaly_score_percentage"
    ],
    label="Anomaly",
    marker="x"
)

plt.title(
    "Anomaly Score by Record"
)

plt.xlabel(
    "Record ID"
)

plt.ylabel(
    "Anomaly Score Percentage"
)

plt.legend()

plt.tight_layout()

plt.show()

if len(feature_columns) >= 2:
    first_feature = feature_columns[0]
    second_feature = feature_columns[1]

    plt.figure(
        figsize=(
            10,
            7
        )
    )

    plt.scatter(
        normal_records[
            first_feature
        ],
        normal_records[
            second_feature
        ],
        label="Normal",
        alpha=0.6
    )

    plt.scatter(
        anomaly_records[
            first_feature
        ],
        anomaly_records[
            second_feature
        ],
        label="Anomaly",
        marker="x"
    )

    plt.title(
        f"Anomaly Detection: "
        f"{first_feature} vs {second_feature}"
    )

    plt.xlabel(
        first_feature
    )

    plt.ylabel(
        second_feature
    )

    plt.legend()

    plt.tight_layout()

    plt.show()

plt.figure(
    figsize=(
        10,
        6
    )
)

plt.barh(
    feature_contribution_df[
        "feature"
    ][::-1],
    feature_contribution_df[
        "contribution_percentage"
    ][::-1]
)

plt.title(
    "Feature Contribution to Anomaly Detection"
)

plt.xlabel(
    "Contribution Percentage"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()

plt.show()

def standardize_anomaly_label(value):
    if pd.isna(value):
        return ""

    value = str(
        value
    ).strip().lower()

    anomaly_values = {
        "1",
        "1.0",
        "yes",
        "true",
        "anomaly",
        "anomalous",
        "outlier",
        "fraud",
        "abnormal",
        "suspicious",
        "irregular"
    }

    normal_values = {
        "0",
        "0.0",
        "no",
        "false",
        "normal",
        "regular",
        "valid",
        "legitimate",
        "non anomaly",
        "non-anomaly",
        "not anomaly"
    }

    if value in anomaly_values:
        return "Anomaly"

    if value in normal_values:
        return "Normal"

    return ""

evaluation_available = False
accuracy = np.nan
precision = np.nan
recall = np.nan
f1 = np.nan
report_df = pd.DataFrame()
confusion_matrix_df = pd.DataFrame()

if target_column is not None:
    model_df[
        "actual_anomaly_status"
    ] = (
        model_df[
            target_column
        ]
        .apply(
            standardize_anomaly_label
        )
    )

    evaluation_subset = model_df[
        model_df[
            "actual_anomaly_status"
        ].isin(
            [
                "Normal",
                "Anomaly"
            ]
        )
    ].copy()

    if (
        len(evaluation_subset) >= 5
        and
        evaluation_subset[
            "actual_anomaly_status"
        ].nunique() >= 2
    ):
        evaluation_available = True

        y_true = evaluation_subset[
            "actual_anomaly_status"
        ]

        y_pred = evaluation_subset[
            "anomaly_status"
        ]

        accuracy = accuracy_score(
            y_true,
            y_pred
        )

        precision = precision_score(
            y_true,
            y_pred,
            pos_label="Anomaly",
            zero_division=0
        )

        recall = recall_score(
            y_true,
            y_pred,
            pos_label="Anomaly",
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            y_pred,
            pos_label="Anomaly",
            zero_division=0
        )

        print(
            "\nAnomaly Detection Evaluation"
        )

        print(
            "Accuracy:",
            round(
                accuracy * 100,
                2
            ),
            "%"
        )

        print(
            "Precision:",
            round(
                precision * 100,
                2
            ),
            "%"
        )

        print(
            "Recall:",
            round(
                recall * 100,
                2
            ),
            "%"
        )

        print(
            "F1-score:",
            round(
                f1 * 100,
                2
            ),
            "%"
        )

        print(
            "\nClassification report:"
        )

        print(
            classification_report(
                y_true,
                y_pred,
                zero_division=0
            )
        )

        report_df = pd.DataFrame(
            classification_report(
                y_true,
                y_pred,
                output_dict=True,
                zero_division=0
            )
        ).transpose().reset_index()

        report_df = report_df.rename(
            columns={
                "index":
                    "class_or_metric"
            }
        )

        class_labels = [
            "Normal",
            "Anomaly"
        ]

        matrix = confusion_matrix(
            y_true,
            y_pred,
            labels=class_labels
        )

        confusion_matrix_df = pd.DataFrame(
            matrix,
            index=class_labels,
            columns=class_labels
        )

        print(
            "\nConfusion matrix:"
        )

        display(
            confusion_matrix_df
        )

        fig, ax = plt.subplots(
            figsize=(
                7,
                6
            )
        )

        matrix_display = ConfusionMatrixDisplay(
            confusion_matrix=matrix,
            display_labels=class_labels
        )

        matrix_display.plot(
            ax=ax,
            values_format="d"
        )

        plt.title(
            "Anomaly Detection Confusion Matrix"
        )

        plt.tight_layout()

        plt.show()

    else:
        print(
            "\nThe existing target column does not "
            "contain enough valid labels for evaluation."
        )

evaluation_df = pd.DataFrame({
    "metric": [
        "model_type",
        "contamination_rate",
        "original_records",
        "usable_records",
        "number_of_features",
        "normal_records",
        "anomaly_records",
        "detected_anomaly_percentage",
        "existing_target_column",
        "supervised_evaluation_available",
        "accuracy",
        "precision",
        "recall",
        "f1_score"
    ],
    "value": [
        "Isolation Forest",
        str(
            contamination_rate
        ),
        str(
            len(anomaly_df)
        ),
        str(
            len(model_df)
        ),
        str(
            len(feature_columns)
        ),
        str(
            normal_count
        ),
        str(
            anomaly_count
        ),
        str(
            round(
                anomaly_percentage,
                2
            )
        ),
        str(
            target_column
            if target_column is not None
            else "Not Available"
        ),
        str(
            evaluation_available
        ),
        (
            str(
                float(accuracy)
            )
            if pd.notna(accuracy)
            else "Not Available"
        ),
        (
            str(
                float(precision)
            )
            if pd.notna(precision)
            else "Not Available"
        ),
        (
            str(
                float(recall)
            )
            if pd.notna(recall)
            else "Not Available"
        ),
        (
            str(
                float(f1)
            )
            if pd.notna(f1)
            else "Not Available"
        )
    ]
})

print(
    "\nModel evaluation summary:"
)

display(
    evaluation_df
)

result_columns = [
    "anomaly_record_id",
    "anomaly_date"
]

optional_identifier_columns = [
    "complaint_id",
    "record_id",
    "interaction_id",
    "text_id",
    "post_id",
    "user_id",
    "citizen_id",
    "transaction_id",
    "department",
    "category",
    "complaint_category",
    "district",
    "state",
    "platform",
    "channel",
    "priority",
    "severity"
]

for column in optional_identifier_columns:
    if (
        column in model_df.columns
        and
        column not in result_columns
    ):
        result_columns.append(
            column
        )

for column in feature_columns:
    if column not in result_columns:
        result_columns.append(
            column
        )

result_columns.extend([
    "anomaly_prediction_code",
    "anomaly_status",
    "anomaly_score",
    "anomaly_score_percentage",
    "anomaly_rank",
    "risk_level"
])

if (
    target_column is not None
    and
    target_column not in result_columns
):
    result_columns.append(
        target_column
    )

if (
    "actual_anomaly_status"
    in model_df.columns
    and
    "actual_anomaly_status"
    not in result_columns
):
    result_columns.append(
        "actual_anomaly_status"
    )

anomaly_results_df = model_df[
    result_columns
].copy()

anomaly_results_df[
    "anomaly_score"
] = (
    anomaly_results_df[
        "anomaly_score"
    ].round(6)
)

print(
    "\nFinal anomaly detection results:"
)

display(
    anomaly_results_df.head(20)
)

model_package = {
    "model":
        anomaly_detection_model,
    "feature_columns":
        feature_columns,
    "date_column":
        date_column,
    "target_column":
        target_column,
    "contamination_rate":
        contamination_rate,
    "training_medians":
        {
            column:
                float(
                    model_df[
                        column
                    ].median()
                )
            for column in feature_columns
        },
    "minimum_anomaly_score":
        float(
            minimum_score
        ),
    "maximum_anomaly_score":
        float(
            maximum_score
        )
}

model_filename = (
    "09_anomaly_detection_model.pkl"
)

joblib.dump(
    model_package,
    model_filename
)

print(
    "\nModel saved successfully:",
    model_filename
)

anomaly_results_df.to_csv(
    "09_anomaly_detection_results.csv",
    index=False
)

top_anomalies_df.to_csv(
    "09_anomaly_detection_top_anomalies.csv",
    index=False
)

evaluation_df.to_csv(
    "09_anomaly_detection_model_evaluation.csv",
    index=False
)

distribution_df.to_csv(
    "09_anomaly_detection_distribution.csv",
    index=False
)

feature_summary_df.to_csv(
    "09_anomaly_detection_feature_summary.csv",
    index=False
)

feature_contribution_df.to_csv(
    "09_anomaly_detection_feature_contribution.csv",
    index=False
)

if not report_df.empty:
    report_df.to_csv(
        "09_anomaly_detection_classification_report.csv",
        index=False
    )

if not confusion_matrix_df.empty:
    confusion_sql_df = (
        confusion_matrix_df
        .reset_index()
        .rename(
            columns={
                "index":
                    "actual_anomaly_status"
            }
        )
    )

    confusion_sql_df.to_csv(
        "09_anomaly_detection_confusion_matrix.csv",
        index=False
    )
else:
    confusion_sql_df = pd.DataFrame()

print(
    "\nCSV files saved successfully."
)

anomaly_results_df.to_sql(
    name="anomaly_detection_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

top_anomalies_df.to_sql(
    name="anomaly_detection_top_anomalies",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

evaluation_df.to_sql(
    name="anomaly_detection_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)

distribution_df.to_sql(
    name="anomaly_detection_distribution",
    con=database_engine,
    if_exists="replace",
    index=False
)

feature_summary_df.to_sql(
    name="anomaly_detection_feature_summary",
    con=database_engine,
    if_exists="replace",
    index=False
)

feature_contribution_df.to_sql(
    name="anomaly_detection_feature_contribution",
    con=database_engine,
    if_exists="replace",
    index=False
)

if not report_df.empty:
    report_df.to_sql(
        name="anomaly_detection_classification_report",
        con=database_engine,
        if_exists="replace",
        index=False
    )

if not confusion_sql_df.empty:
    confusion_sql_df.to_sql(
        name="anomaly_detection_confusion_matrix",
        con=database_engine,
        if_exists="replace",
        index=False
    )

print(
    "\nAll Anomaly Detection tables "
    "created in MySQL."
)

def predict_anomaly(new_record):
    if not isinstance(
        new_record,
        dict
    ):
        raise TypeError(
            "Enter the new record as a Python dictionary."
        )

    prepared_record = {}

    for column in feature_columns:
        default_value = model_package[
            "training_medians"
        ][column]

        value = new_record.get(
            column,
            default_value
        )

        try:
            value = float(
                value
            )
        except (
            TypeError,
            ValueError
        ):
            value = default_value

        if not np.isfinite(
            value
        ):
            value = default_value

        prepared_record[column] = value

    input_df = pd.DataFrame(
        [
            prepared_record
        ]
    )

    prediction_code = int(
        anomaly_detection_model.predict(
            input_df
        )[0]
    )

    decision_score = float(
        anomaly_detection_model.decision_function(
            input_df
        )[0]
    )

    anomaly_score = (
        -decision_score
    )

    minimum_training_score = (
        model_package[
            "minimum_anomaly_score"
        ]
    )

    maximum_training_score = (
        model_package[
            "maximum_anomaly_score"
        ]
    )

    if (
        maximum_training_score
        >
        minimum_training_score
    ):
        anomaly_score_percentage = (
            (
                anomaly_score
                -
                minimum_training_score
            )
            /
            (
                maximum_training_score
                -
                minimum_training_score
            )
            *
            100
        )
    else:
        anomaly_score_percentage = 0

    anomaly_score_percentage = max(
        0,
        min(
            100,
            anomaly_score_percentage
        )
    )

    predicted_status = (
        "Anomaly"
        if prediction_code == -1
        else "Normal"
    )

    if anomaly_score_percentage <= 25:
        risk_level = "Low"

    elif anomaly_score_percentage <= 50:
        risk_level = "Medium"

    elif anomaly_score_percentage <= 75:
        risk_level = "High"

    else:
        risk_level = "Critical"

    print(
        "\nInput record:"
    )

    display(
        input_df
    )

    print(
        "Predicted anomaly status:",
        predicted_status
    )

    print(
        "Anomaly score:",
        round(
            anomaly_score,
            6
        )
    )

    print(
        "Anomaly score percentage:",
        round(
            anomaly_score_percentage,
            2
        ),
        "%"
    )

    print(
        "Risk level:",
        risk_level
    )

    return {
        "input_record":
            prepared_record,
        "predicted_anomaly_status":
            predicted_status,
        "anomaly_score":
            round(
                anomaly_score,
                6
            ),
        "anomaly_score_percentage":
            round(
                anomaly_score_percentage,
                2
            ),
        "risk_level":
            risk_level
    }

normal_sample = {
    column:
        float(
            model_df[
                column
            ].median()
        )
    for column in feature_columns
}

predict_anomaly(
    normal_sample
)

high_value_sample = {
    column:
        float(
            model_df[
                column
            ].median()
        )
    for column in feature_columns
}

for column in feature_columns:
    column_std = model_df[
        column
    ].std()

    if pd.isna(
        column_std
    ) or column_std == 0:
        column_std = 1

    high_value_sample[column] = (
        model_df[
            column
        ].median()
        +
        10
        *
        column_std
    )

predict_anomaly(
    high_value_sample
)

mysql_table_names = [
    "anomaly_detection_results",
    "anomaly_detection_top_anomalies",
    "anomaly_detection_model_evaluation",
    "anomaly_detection_distribution",
    "anomaly_detection_feature_summary",
    "anomaly_detection_feature_contribution"
]

mysql_table_rows = [
    len(
        anomaly_results_df
    ),
    len(
        top_anomalies_df
    ),
    len(
        evaluation_df
    ),
    len(
        distribution_df
    ),
    len(
        feature_summary_df
    ),
    len(
        feature_contribution_df
    )
]

if not report_df.empty:
    mysql_table_names.append(
        "anomaly_detection_classification_report"
    )

    mysql_table_rows.append(
        len(
            report_df
        )
    )

if not confusion_sql_df.empty:
    mysql_table_names.append(
        "anomaly_detection_confusion_matrix"
    )

    mysql_table_rows.append(
        len(
            confusion_sql_df
        )
    )

verification_df = pd.DataFrame({
    "table_name":
        mysql_table_names,
    "total_rows":
        mysql_table_rows
})

print(
    "\nMySQL tables created:"
)

display(
    verification_df
)

sql_anomaly_sample = pd.read_sql(
    """
    SELECT *
    FROM anomaly_detection_results
    ORDER BY anomaly_score DESC
    LIMIT 10
    """,
    database_engine
)

print(
    "\nTop anomaly results read from MySQL:"
)

display(
    sql_anomaly_sample
)

print(
    "\n09_Anomaly_Detection "
    "completed successfully."
)

print(
    "\nSource table:",
    source_table
)

print(
    "Selected date column:",
    date_column
)

print(
    "Selected numeric features:"
)

for index, column in enumerate(
    feature_columns,
    start=1
):
    print(
        f"{index}. {column}"
    )

print(
    "\nOriginal records:",
    len(anomaly_df)
)

print(
    "Usable records:",
    len(model_df)
)

print(
    "Normal records:",
    normal_count
)

print(
    "Anomaly records:",
    anomaly_count
)

print(
    "Anomaly percentage:",
    round(
        anomaly_percentage,
        2
    ),
    "%"
)

print(
    "Contamination rate:",
    contamination_rate
)

if evaluation_available:
    print(
        "Accuracy:",
        round(
            accuracy * 100,
            2
        ),
        "%"
    )

    print(
        "F1-score:",
        round(
            f1 * 100,
            2
        ),
        "%"
    )

print(
    "\nSaved model:",
    model_filename
)

print(
    "\nCSV output files:"
)

print(
    "1. 09_anomaly_detection_results.csv"
)

print(
    "2. 09_anomaly_detection_top_anomalies.csv"
)

print(
    "3. 09_anomaly_detection_model_evaluation.csv"
)

print(
    "4. 09_anomaly_detection_distribution.csv"
)

print(
    "5. 09_anomaly_detection_feature_summary.csv"
)

print(
    "6. 09_anomaly_detection_feature_contribution.csv"
)

if not report_df.empty:
    print(
        "7. 09_anomaly_detection_classification_report.csv"
    )

if not confusion_sql_df.empty:
    print(
        "8. 09_anomaly_detection_confusion_matrix.csv"
    )

print(
    "\nMySQL output tables:"
)

for index, table_name in enumerate(
    mysql_table_names,
    start=1
):
    print(
        f"{index}. {table_name}"
    )
```
