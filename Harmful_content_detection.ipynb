```python
import os
import re
import getpass
import joblib
import warnings
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"
database_name = "government_analytics"
source_table = "harmful_content_detection"

mysql_password = getpass.getpass("Enter MySQL password: ")
encoded_password = quote_plus(mysql_password)

server_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)

with server_engine.begin() as connection:
    connection.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS `{database_name}`"
        )
    )

database_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/{database_name}",
    pool_pre_ping=True
)

with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()

print("Connected database:", connected_database)

existing_tables = inspect(database_engine).get_table_names()

if source_table in existing_tables:
    harmful_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print("Dataset loaded from MySQL.")

else:
    possible_paths = [
        "harmful_content_detection.csv",
        "harmful_content_detection(1).csv",
        "harmful_content_detection(2).csv",
        "harmful_content.csv",
        "harmful_content(1).csv",
        "harmful_content(2).csv",
        "07_Harmful_Content_Detection.csv",
        "/mnt/data/harmful_content_detection.csv",
        "/mnt/data/harmful_content_detection(1).csv",
        "/mnt/data/harmful_content_detection(2).csv",
        "/mnt/data/harmful_content.csv",
        "/mnt/data/harmful_content(1).csv",
        "/mnt/data/harmful_content(2).csv",
        "/mnt/data/07_Harmful_Content_Detection.csv",
        r"C:\Users\Kishore\Downloads\harmful_content_detection.csv",
        r"C:\Users\Kishore\Downloads\harmful_content_detection(1).csv",
        r"C:\Users\Kishore\Downloads\harmful_content_detection(2).csv",
        r"C:\Users\Kishore\Downloads\harmful_content.csv",
        r"C:\Users\Kishore\Downloads\harmful_content(1).csv",
        r"C:\Users\Kishore\Downloads\harmful_content(2).csv",
        r"C:\Users\Kishore\Downloads\07_Harmful_Content_Detection.csv"
    ]

    csv_path = next(
        (
            path
            for path in possible_paths
            if os.path.exists(path)
        ),
        None
    )

    if csv_path is None:
        raise FileNotFoundError(
            "Place your harmful-content CSV file in the same "
            "folder as the Jupyter Notebook."
        )

    harmful_df = pd.read_csv(
        csv_path,
        low_memory=False
    )

    harmful_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print("Dataset loaded from CSV:", csv_path)
    print("Dataset uploaded to MySQL.")

if harmful_df.empty:
    raise ValueError("The dataset is empty.")

print("\nOriginal dataset shape:", harmful_df.shape)

print("\nOriginal columns:")
print(harmful_df.columns.tolist())

display(harmful_df.head())

harmful_df.columns = (
    harmful_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(
        r"[^a-z0-9]+",
        "_",
        regex=True
    )
    .str.strip("_")
)

unnamed_columns = [
    column
    for column in harmful_df.columns
    if column.startswith("unnamed")
]

harmful_df = harmful_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)

print("\nStandardized columns:")
print(harmful_df.columns.tolist())

text_candidates = [
    "text",
    "content_text",
    "message",
    "post_text",
    "comment_text",
    "tweet_text",
    "social_media_text",
    "interaction_summary",
    "complaint_text",
    "complaint_description",
    "feedback_text",
    "description",
    "content",
    "statement",
    "summary"
]

text_column = next(
    (
        column
        for column in text_candidates
        if column in harmful_df.columns
    ),
    None
)

if text_column is None:
    object_columns = harmful_df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    object_columns = [
        column
        for column in object_columns
        if "id" not in column.lower()
        and not column.lower().startswith("task")
    ]

    if object_columns:
        text_column = max(
            object_columns,
            key=lambda column: (
                harmful_df[column]
                .fillna("")
                .astype(str)
                .str.len()
                .mean()
            )
        )

if text_column is None:
    raise ValueError(
        "No text column was found. Available columns: "
        f"{harmful_df.columns.tolist()}"
    )

def basic_label_cleaning(value):
    if pd.isna(value):
        return ""

    value = str(value).strip().lower()

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value

invalid_target_values = {
    "",
    "nan",
    "none",
    "null",
    "n/a",
    "na",
    "not available",
    "unknown",
    "unlabelled",
    "unlabeled",
    "-"
}

target_candidates = [
    "harmful_label",
    "harmful_content_label",
    "harmful_content",
    "is_harmful",
    "harmful",
    "toxicity_label",
    "toxicity",
    "toxic",
    "content_label",
    "moderation_label",
    "safety_label",
    "label",
    "target",
    "class",
    "task_1",
    "task_2",
    "task_3"
]

available_target_candidates = [
    column
    for column in target_candidates
    if column in harmful_df.columns
]

def target_column_is_usable(dataframe, column):
    cleaned_values = (
        dataframe[column]
        .apply(basic_label_cleaning)
    )

    cleaned_values = cleaned_values[
        ~cleaned_values.isin(invalid_target_values)
    ]

    if cleaned_values.empty:
        return False

    counts = cleaned_values.value_counts()

    if len(counts) < 2:
        return False

    if len(counts) > 30:
        return False

    if counts.min() < 2:
        return False

    return True

target_column = None

for candidate in available_target_candidates:
    if target_column_is_usable(
        harmful_df,
        candidate
    ):
        target_column = candidate
        break

if target_column is None:
    possible_target_columns = [
        column
        for column in harmful_df.columns
        if column != text_column
        and "id" not in column.lower()
    ]

    for candidate in possible_target_columns:
        if target_column_is_usable(
            harmful_df,
            candidate
        ):
            target_column = candidate
            break

if target_column is None:
    raise ValueError(
        "No usable target column was found. "
        "The columns task_1, task_2 and task_3 must contain "
        "at least two different labels. Available columns: "
        f"{harmful_df.columns.tolist()}"
    )

print("\nInput text column:", text_column)
print("Selected target column:", target_column)

print("\nSelected target original values:")

display(
    harmful_df[target_column]
    .value_counts(dropna=False)
    .rename_axis("original_label")
    .reset_index(name="record_count")
)

harmful_df[text_column] = (
    harmful_df[text_column]
    .fillna("")
    .astype(str)
    .str.strip()
)

harmful_df[target_column] = (
    harmful_df[target_column]
    .fillna("")
    .astype(str)
    .str.strip()
)

harmful_df["content_text"] = (
    harmful_df[text_column]
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
    .str.strip()
)

def standardize_harmful_label(value):
    value = basic_label_cleaning(value)

    harmful_values = {
        "1",
        "1.0",
        "yes",
        "y",
        "true",
        "harmful",
        "toxic",
        "abusive",
        "abuse",
        "offensive",
        "offence",
        "hate",
        "hate speech",
        "hateful",
        "threat",
        "threatening",
        "violent",
        "violence",
        "unsafe",
        "cyberbullying",
        "bullying",
        "harassment",
        "harassing",
        "spam harmful",
        "malicious",
        "inappropriate"
    }

    safe_values = {
        "0",
        "0.0",
        "no",
        "n",
        "false",
        "safe",
        "non harmful",
        "non-harmful",
        "not harmful",
        "nonharmful",
        "non toxic",
        "non-toxic",
        "not toxic",
        "nontoxic",
        "normal",
        "clean",
        "acceptable",
        "benign",
        "appropriate",
        "not offensive",
        "non offensive",
        "non-offensive",
        "not abusive",
        "not hateful",
        "no hate"
    }

    if value in harmful_values:
        return "Harmful"

    if value in safe_values:
        return "Safe"

    if value in invalid_target_values:
        return ""

    return value.title()

harmful_df["harmful_label"] = (
    harmful_df[target_column]
    .apply(standardize_harmful_label)
)

print("\nStandardized target values:")

display(
    harmful_df["harmful_label"]
    .value_counts(dropna=False)
    .rename_axis("harmful_label")
    .reset_index(name="record_count")
)

def remove_harmful_label_leakage(value):
    value = str(value)

    leakage_patterns = [
        (
            r"\bharmful\s*[:=\-]?\s*"
            r"(yes|no|true|false|1|0)\b"
        ),
        (
            r"\bis[_\s-]*harmful\s*[:=\-]?\s*"
            r"(yes|no|true|false|1|0)\b"
        ),
        (
            r"\btoxicity\s*[:=\-]?\s*"
            r"(toxic|non[-\s]?toxic|yes|no|1|0)\b"
        ),
        (
            r"\blabel\s*[:=\-]?\s*"
            r"(harmful|safe|toxic|non[-\s]?toxic)\b"
        ),
        (
            r"\btask[_\s-]*1\s*[:=\-]?\s*"
            r"(harmful|safe|toxic|non[-\s]?toxic|yes|no|1|0)\b"
        ),
        (
            r"\btask[_\s-]*2\s*[:=\-]?\s*"
            r"(harmful|safe|toxic|non[-\s]?toxic|yes|no|1|0)\b"
        ),
        (
            r"\btask[_\s-]*3\s*[:=\-]?\s*"
            r"(harmful|safe|toxic|non[-\s]?toxic|yes|no|1|0)\b"
        )
    ]

    for pattern in leakage_patterns:
        value = re.sub(
            pattern,
            " ",
            value,
            flags=re.IGNORECASE
        )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()

harmful_df["content_text"] = (
    harmful_df["content_text"]
    .apply(remove_harmful_label_leakage)
)

def clean_content_text(value):
    value = str(value).lower()

    value = re.sub(
        r"http\S+|www\S+",
        " ",
        value
    )

    value = re.sub(
        r"\S+@\S+",
        " ",
        value
    )

    value = re.sub(
        r"@\w+",
        " ",
        value
    )

    value = re.sub(
        r"#(\w+)",
        r"\1",
        value
    )

    value = re.sub(
        r"&[a-z]+;",
        " ",
        value
    )

    value = re.sub(
        r"[^a-z\s]",
        " ",
        value
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()

harmful_df["clean_text"] = (
    harmful_df["content_text"]
    .apply(clean_content_text)
)

print("\nCleaned data sample:")

display(
    harmful_df[
        [
            text_column,
            target_column,
            "content_text",
            "clean_text",
            "harmful_label"
        ]
    ].head(10)
)

selected_columns = [
    "content_text",
    "clean_text",
    "harmful_label"
]

optional_columns = [
    "text_id",
    "content_id",
    "post_id",
    "comment_id",
    "tweet_id",
    "interaction_id",
    "citizen_id",
    "user_id",
    "timestamp",
    "date",
    "platform",
    "channel",
    "department",
    "scheme_name",
    "priority",
    "state",
    "district",
    "language",
    "sentiment_score",
    "sentiment_label",
    "severity",
    "reported",
    "moderation_status",
    "resolution_time_hrs",
    "escalated",
    "task_1",
    "task_2",
    "task_3"
]

for column in optional_columns:
    if (
        column in harmful_df.columns
        and column not in selected_columns
    ):
        selected_columns.append(column)

model_df = harmful_df[
    selected_columns
].copy()

model_df = model_df.dropna(
    subset=[
        "clean_text",
        "harmful_label"
    ]
)

model_df = model_df[
    model_df["clean_text"].str.len() >= 3
].copy()

model_df = model_df[
    ~model_df["harmful_label"]
    .str.lower()
    .isin(invalid_target_values)
].copy()

model_df = model_df.drop_duplicates(
    subset=[
        "clean_text",
        "harmful_label"
    ]
).reset_index(drop=True)

class_counts_before = (
    model_df["harmful_label"]
    .value_counts()
)

print("\nClass distribution before filtering:")

display(
    class_counts_before
    .rename_axis("harmful_status")
    .reset_index(name="record_count")
)

minimum_records_per_class = 2

valid_classes = class_counts_before[
    class_counts_before >= minimum_records_per_class
].index

model_df = model_df[
    model_df["harmful_label"].isin(
        valid_classes
    )
].reset_index(drop=True)

if len(model_df) < 10:
    raise ValueError(
        "Not enough usable records are available after cleaning. "
        "At least 10 records are required."
    )

if model_df["harmful_label"].nunique() < 2:
    raise ValueError(
        "At least two target classes are required. "
        f"Selected target column: {target_column}"
    )

final_class_counts = (
    model_df["harmful_label"]
    .value_counts()
)

if final_class_counts.min() < 2:
    raise ValueError(
        "Every target class must contain at least two records."
    )

print("\nFinal class distribution:")

display(
    final_class_counts
    .rename_axis("harmful_status")
    .reset_index(name="record_count")
)

print("\nOriginal records:", len(harmful_df))
print("Usable records:", len(model_df))

print(
    "Number of classes:",
    model_df["harmful_label"].nunique()
)

X = model_df["clean_text"]
y = model_df["harmful_label"]

number_of_classes = y.nunique()
number_of_records = len(model_df)

minimum_test_records = number_of_classes
default_test_records = int(
    round(number_of_records * 0.20)
)

test_record_count = max(
    minimum_test_records,
    default_test_records
)

maximum_test_records = (
    number_of_records - number_of_classes
)

test_record_count = min(
    test_record_count,
    maximum_test_records
)

if test_record_count < number_of_classes:
    raise ValueError(
        "The dataset does not contain enough records "
        "for a stratified train-test split."
    )

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=test_record_count,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

harmful_content_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            stop_words=None,
            max_features=20000,
            ngram_range=(1, 2),
            min_df=1,
            max_df=1.0,
            sublinear_tf=True,
            token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            solver="liblinear",
            random_state=42
        )
    )
])

harmful_content_model.fit(
    X_train,
    y_train
)

print(
    "\nHarmful Content Detection model "
    "trained successfully."
)

y_pred = harmful_content_model.predict(
    X_test
)

prediction_probabilities = (
    harmful_content_model.predict_proba(
        X_test
    )
)

prediction_confidence = (
    prediction_probabilities.max(axis=1)
)

accuracy = accuracy_score(
    y_test,
    y_pred
)

weighted_precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

print("\nHarmful Content Detection Model Evaluation")

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Weighted precision:",
    round(weighted_precision * 100, 2),
    "%"
)

print(
    "Weighted recall:",
    round(weighted_recall * 100, 2),
    "%"
)

print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)

print(
    "Macro F1-score:",
    round(macro_f1 * 100, 2),
    "%"
)

print("\nClassification report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

report_df = pd.DataFrame(
    classification_report(
        y_test,
        y_pred,
        output_dict=True,
        zero_division=0
    )
).transpose().reset_index()

report_df = report_df.rename(
    columns={
        "index": "class_or_metric"
    }
)

display(report_df)

class_labels = sorted(
    set(y_test.unique()).union(
        set(y_pred)
    )
)

matrix = confusion_matrix(
    y_test,
    y_pred,
    labels=class_labels
)

confusion_matrix_df = pd.DataFrame(
    matrix,
    index=class_labels,
    columns=class_labels
)

print("\nConfusion matrix:")

display(confusion_matrix_df)

fig, ax = plt.subplots(
    figsize=(10, 8)
)

matrix_display = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_labels
)

matrix_display.plot(
    ax=ax,
    values_format="d",
    xticks_rotation=45
)

plt.title(
    "Harmful Content Detection Confusion Matrix"
)

plt.tight_layout()
plt.show()

test_indices = X_test.index

prediction_results = pd.DataFrame({
    "content_text": model_df.loc[
        test_indices,
        "content_text"
    ].values,
    "clean_text": X_test.values,
    "actual_harmful_status": y_test.values,
    "predicted_harmful_status": y_pred,
    "prediction_confidence": prediction_confidence
})

prediction_results[
    "prediction_confidence_percentage"
] = (
    prediction_results[
        "prediction_confidence"
    ] * 100
).round(2)

prediction_results[
    "prediction_correct"
] = (
    prediction_results[
        "actual_harmful_status"
    ]
    ==
    prediction_results[
        "predicted_harmful_status"
    ]
)

prediction_results.insert(
    0,
    "prediction_id",
    range(
        1,
        len(prediction_results) + 1
    )
)

print("\nPrediction results:")

display(
    prediction_results.head(20)
)

evaluation_df = pd.DataFrame({
    "metric": [
        "accuracy",
        "weighted_precision",
        "weighted_recall",
        "weighted_f1_score",
        "macro_f1_score",
        "original_records",
        "usable_records",
        "training_records",
        "testing_records",
        "number_of_classes",
        "selected_target_column"
    ],
    "value": [
        str(float(accuracy)),
        str(float(weighted_precision)),
        str(float(weighted_recall)),
        str(float(weighted_f1)),
        str(float(macro_f1)),
        str(int(len(harmful_df))),
        str(int(len(model_df))),
        str(int(len(X_train))),
        str(int(len(X_test))),
        str(int(y.nunique())),
        str(target_column)
    ]
})

print("\nModel evaluation summary:")

display(evaluation_df)

labelled_data = model_df.copy()

labelled_data.insert(
    0,
    "labelled_record_id",
    range(
        1,
        len(labelled_data) + 1
    )
)

distribution_df = (
    model_df["harmful_label"]
    .value_counts()
    .rename_axis("harmful_status")
    .reset_index(name="record_count")
)

confusion_sql_df = (
    confusion_matrix_df
    .reset_index()
    .rename(
        columns={
            "index": "actual_harmful_status"
        }
    )
)

model_package = {
    "model": harmful_content_model,
    "text_column": text_column,
    "target_column": target_column,
    "classes": harmful_content_model.classes_.tolist()
}

model_filename = (
    "07_harmful_content_detection_model.pkl"
)

joblib.dump(
    model_package,
    model_filename
)

print(
    "\nModel saved successfully:",
    model_filename
)

prediction_results.to_csv(
    "07_harmful_content_detection_results.csv",
    index=False
)

evaluation_df.to_csv(
    "07_harmful_content_model_evaluation.csv",
    index=False
)

report_df.to_csv(
    "07_harmful_content_classification_report.csv",
    index=False
)

labelled_data.to_csv(
    "07_harmful_content_labelled_data.csv",
    index=False
)

distribution_df.to_csv(
    "07_harmful_content_class_distribution.csv",
    index=False
)

confusion_sql_df.to_csv(
    "07_harmful_content_confusion_matrix.csv",
    index=False
)

print("\nCSV files saved successfully.")

prediction_results.to_sql(
    name="harmful_content_detection_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

evaluation_df.to_sql(
    name="harmful_content_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)

report_df.to_sql(
    name="harmful_content_classification_report",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

labelled_data.to_sql(
    name="harmful_content_labelled_data",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

distribution_df.to_sql(
    name="harmful_content_class_distribution",
    con=database_engine,
    if_exists="replace",
    index=False
)

confusion_sql_df.to_sql(
    name="harmful_content_confusion_matrix",
    con=database_engine,
    if_exists="replace",
    index=False
)

print(
    "\nAll Harmful Content Detection "
    "tables created in MySQL."
)

def predict_harmful_content(new_content):
    if not isinstance(new_content, str):
        raise TypeError(
            "The content must be entered as text."
        )

    cleaned_content = remove_harmful_label_leakage(
        new_content
    )

    cleaned_content = clean_content_text(
        cleaned_content
    )

    if len(cleaned_content) < 3:
        raise ValueError(
            "Enter meaningful text for prediction."
        )

    predicted_status = (
        harmful_content_model.predict(
            [cleaned_content]
        )[0]
    )

    probabilities = (
        harmful_content_model.predict_proba(
            [cleaned_content]
        )[0]
    )

    probability_df = pd.DataFrame({
        "harmful_status":
            harmful_content_model.classes_,
        "probability_percentage":
            (
                probabilities * 100
            ).round(2)
    })

    probability_df = (
        probability_df
        .sort_values(
            by="probability_percentage",
            ascending=False
        )
        .reset_index(drop=True)
    )

    confidence = float(
        probability_df.iloc[0][
            "probability_percentage"
        ]
    )

    print("\nContent:", new_content)

    print(
        "Predicted status:",
        predicted_status
    )

    print(
        "Prediction confidence:",
        confidence,
        "%"
    )

    print("\nClass probabilities:")

    display(probability_df)

    return {
        "content": new_content,
        "predicted_harmful_status":
            predicted_status,
        "confidence_percentage":
            confidence
    }

predict_harmful_content(
    "Thank you for providing this useful government service."
)

predict_harmful_content(
    "You are useless and nobody wants you here."
)

predict_harmful_content(
    "The government website is working correctly now."
)

predict_harmful_content(
    "I will hurt you if you continue doing this."
)

verification_df = pd.DataFrame({
    "table_name": [
        "harmful_content_detection_results",
        "harmful_content_model_evaluation",
        "harmful_content_classification_report",
        "harmful_content_labelled_data",
        "harmful_content_class_distribution",
        "harmful_content_confusion_matrix"
    ],
    "total_rows": [
        len(prediction_results),
        len(evaluation_df),
        len(report_df),
        len(labelled_data),
        len(distribution_df),
        len(confusion_sql_df)
    ]
})

print("\nMySQL tables created:")

display(verification_df)

sql_prediction_sample = pd.read_sql(
    """
    SELECT *
    FROM harmful_content_detection_results
    LIMIT 10
    """,
    database_engine
)

print("\nSample predictions read from MySQL:")

display(sql_prediction_sample)

print(
    "\n07_Harmful_Content_Detection "
    "completed successfully."
)

print("\nSource table:", source_table)
print("Input text column:", text_column)
print("Selected target column:", target_column)
print("Original records:", len(harmful_df))
print("Usable records:", len(model_df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Number of classes:", y.nunique())

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)

print("\nSaved model:", model_filename)

print("\nCSV output files:")
print("1. 07_harmful_content_detection_results.csv")
print("2. 07_harmful_content_model_evaluation.csv")
print("3. 07_harmful_content_classification_report.csv")
print("4. 07_harmful_content_labelled_data.csv")
print("5. 07_harmful_content_class_distribution.csv")
print("6. 07_harmful_content_confusion_matrix.csv")

print("\nMySQL output tables:")
print("1. harmful_content_detection_results")
print("2. harmful_content_model_evaluation")
print("3. harmful_content_classification_report")
print("4. harmful_content_labelled_data")
print("5. harmful_content_class_distribution")
print("6. harmful_content_confusion_matrix")
```
