```python
import os
import re
import getpass
import joblib
import warnings
import pandas as pd
import matplotlib.pyplot as plt

from urllib.parse import quote_plus
from sqlalchemy import create_engine, text, inspect

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

mysql_username = "root"
mysql_host = "localhost"
mysql_port = "3306"
database_name = "government_analytics"
source_table = "enterprise_department_priority"

mysql_password = getpass.getpass("Enter MySQL password: ")
encoded_password = quote_plus(mysql_password)

server_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}",
    pool_pre_ping=True
)

with server_engine.begin() as connection:
    connection.execute(
        text(
            f"CREATE DATABASE IF NOT EXISTS `{database_name}`"
        )
    )

database_engine = create_engine(
    f"mysql+mysqlconnector://{mysql_username}:{encoded_password}"
    f"@{mysql_host}:{mysql_port}/{database_name}",
    pool_pre_ping=True
)

with database_engine.connect() as connection:
    connected_database = connection.execute(
        text("SELECT DATABASE()")
    ).scalar()

print("Connected database:", connected_database)

existing_tables = inspect(database_engine).get_table_names()

if source_table in existing_tables:
    department_df = pd.read_sql(
        f"SELECT * FROM `{source_table}`",
        database_engine
    )

    print("Dataset loaded from MySQL.")

else:
    possible_paths = [
        "enterprise_department_priority(2).csv",
        "enterprise_department_priority.csv",
        "/mnt/data/enterprise_department_priority(2).csv",
        r"C:\Users\Kishore\Downloads\enterprise_department_priority(2).csv",
        r"C:\Users\Kishore\Downloads\enterprise_department_priority.csv"
    ]

    csv_path = next(
        (
            path
            for path in possible_paths
            if os.path.exists(path)
        ),
        None
    )

    if csv_path is None:
        raise FileNotFoundError(
            "Place enterprise_department_priority(2).csv "
            "in the same folder as the notebook."
        )

    department_df = pd.read_csv(
        csv_path,
        low_memory=False
    )

    department_df.to_sql(
        name=source_table,
        con=database_engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

    print("Dataset loaded from CSV and uploaded to MySQL.")

if department_df.empty:
    raise ValueError("The dataset is empty.")

print("\nOriginal dataset shape:", department_df.shape)

print("\nOriginal columns:")
print(department_df.columns.tolist())

display(department_df.head())

department_df.columns = (
    department_df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

unnamed_columns = [
    column
    for column in department_df.columns
    if column.startswith("unnamed")
]

department_df = department_df.drop(
    columns=unnamed_columns,
    errors="ignore"
)

text_candidates = [
    "interaction_summary",
    "complaint_text",
    "complaint_description",
    "description",
    "feedback_text",
    "text",
    "message",
    "summary"
]

target_candidates = [
    "department",
    "department_name",
    "assigned_department",
    "ministry",
    "ministry_name"
]

text_column = next(
    (
        column
        for column in text_candidates
        if column in department_df.columns
    ),
    None
)

target_column = next(
    (
        column
        for column in target_candidates
        if column in department_df.columns
    ),
    None
)

if text_column is None:
    raise ValueError(
        "No interaction or complaint-text column was found. "
        f"Available columns: {department_df.columns.tolist()}"
    )

if target_column is None:
    raise ValueError(
        "No department target column was found. "
        f"Available columns: {department_df.columns.tolist()}"
    )

print("\nInput text column:", text_column)
print("Target department column:", target_column)

department_df[text_column] = (
    department_df[text_column]
    .fillna("")
    .astype(str)
    .str.strip()
)

department_df[target_column] = (
    department_df[target_column]
    .fillna("")
    .astype(str)
    .str.strip()
)

department_df["interaction_text"] = (
    department_df[text_column]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

department_df["department_label"] = (
    department_df[target_column]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

def clean_interaction_text(value):
    value = str(value).lower()
    value = re.sub(r"http\S+|www\S+", " ", value)
    value = re.sub(r"\S+@\S+", " ", value)
    value = re.sub(r"@\w+", " ", value)
    value = re.sub(r"#(\w+)", r"\1", value)
    value = re.sub(r"&[a-z]+;", " ", value)
    value = re.sub(r"[^a-z\s]", " ", value)
    value = re.sub(r"\s+", " ", value)
    return value.strip()

department_df["clean_text"] = (
    department_df["interaction_text"]
    .apply(clean_interaction_text)
)

selected_columns = [
    "interaction_text",
    "clean_text",
    "department_label"
]

optional_columns = [
    "interaction_id",
    "citizen_id",
    "timestamp",
    "channel",
    "scheme_name",
    "priority",
    "state",
    "language",
    "sentiment_score",
    "resolution_time_hrs",
    "reopen_count",
    "escalated",
    "satisfaction_rating",
    "cost_to_resolve_inr",
    "complexity_index",
    "sentiment_label"
]

for column in optional_columns:
    if column in department_df.columns:
        selected_columns.append(column)

model_df = department_df[
    selected_columns
].copy()

model_df = model_df.dropna(
    subset=[
        "clean_text",
        "department_label"
    ]
)

invalid_labels = [
    "",
    "nan",
    "none",
    "null",
    "n/a",
    "na",
    "not available",
    "unknown",
    "-"
]

model_df = model_df[
    model_df["clean_text"].str.len() >= 5
].copy()

model_df = model_df[
    ~model_df["department_label"]
    .str.lower()
    .isin(invalid_labels)
].copy()

model_df = model_df.drop_duplicates(
    subset=[
        "clean_text",
        "department_label"
    ]
).reset_index(drop=True)

department_counts_before = (
    model_df["department_label"]
    .value_counts()
)

print("\nDepartment distribution before filtering:")

display(
    department_counts_before
    .rename_axis("department")
    .reset_index(name="record_count")
)

minimum_records_per_department = 10

valid_departments = department_counts_before[
    department_counts_before >= minimum_records_per_department
].index

model_df = model_df[
    model_df["department_label"].isin(
        valid_departments
    )
].reset_index(drop=True)

if len(model_df) < 20:
    raise ValueError(
        "Not enough usable records are available."
    )

if model_df["department_label"].nunique() < 2:
    raise ValueError(
        "At least two departments are required "
        "for classification."
    )

minimum_department_count = (
    model_df["department_label"]
    .value_counts()
    .min()
)

if minimum_department_count < 2:
    raise ValueError(
        "Each department must contain at least two records."
    )

print("\nFinal department distribution:")

display(
    model_df["department_label"]
    .value_counts()
    .rename_axis("department")
    .reset_index(name="record_count")
)

print("\nOriginal records:", len(department_df))
print("Usable records:", len(model_df))
print(
    "Number of departments:",
    model_df["department_label"].nunique()
)

X = model_df["clean_text"]
y = model_df["department_label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

department_prediction_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            stop_words="english",
            max_features=15000,
            ngram_range=(1, 2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z]+\b"
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            solver="lbfgs",
            random_state=42
        )
    )
])

department_prediction_model.fit(
    X_train,
    y_train
)

print("\nDepartment prediction model trained successfully.")

y_pred = department_prediction_model.predict(
    X_test
)

prediction_probabilities = (
    department_prediction_model.predict_proba(
        X_test
    )
)

prediction_confidence = (
    prediction_probabilities.max(axis=1)
)

accuracy = accuracy_score(
    y_test,
    y_pred
)

weighted_precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

print("\nDepartment Prediction Model Evaluation")

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Weighted precision:",
    round(weighted_precision * 100, 2),
    "%"
)

print(
    "Weighted recall:",
    round(weighted_recall * 100, 2),
    "%"
)

print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)

print(
    "Macro F1-score:",
    round(macro_f1 * 100, 2),
    "%"
)

print("\nClassification report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

report_df = pd.DataFrame(
    classification_report(
        y_test,
        y_pred,
        output_dict=True,
        zero_division=0
    )
).transpose().reset_index()

report_df = report_df.rename(
    columns={
        "index": "class_or_metric"
    }
)

display(report_df)

class_labels = sorted(
    set(y_test.unique()).union(
        set(y_pred)
    )
)

matrix = confusion_matrix(
    y_test,
    y_pred,
    labels=class_labels
)

confusion_matrix_df = pd.DataFrame(
    matrix,
    index=class_labels,
    columns=class_labels
)

print("\nConfusion matrix:")

display(confusion_matrix_df)

fig, ax = plt.subplots(
    figsize=(12, 10)
)

display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_labels
)

display_matrix.plot(
    ax=ax,
    values_format="d",
    xticks_rotation=90
)

plt.title(
    "Department Prediction Confusion Matrix"
)

plt.tight_layout()
plt.show()

prediction_results = pd.DataFrame({
    "interaction_text": X_test.values,
    "actual_department": y_test.values,
    "predicted_department": y_pred,
    "prediction_confidence": prediction_confidence
})

prediction_results[
    "prediction_confidence_percentage"
] = (
    prediction_results[
        "prediction_confidence"
    ] * 100
).round(2)

prediction_results[
    "prediction_correct"
] = (
    prediction_results[
        "actual_department"
    ]
    == prediction_results[
        "predicted_department"
    ]
)

prediction_results.insert(
    0,
    "prediction_id",
    range(
        1,
        len(prediction_results) + 1
    )
)

print("\nDepartment prediction results:")

display(
    prediction_results.head(20)
)

evaluation_df = pd.DataFrame({
    "metric": [
        "accuracy",
        "weighted_precision",
        "weighted_recall",
        "weighted_f1_score",
        "macro_f1_score",
        "original_records",
        "usable_records",
        "training_records",
        "testing_records",
        "number_of_departments",
        "minimum_records_per_department"
    ],
    "value": [
        float(accuracy),
        float(weighted_precision),
        float(weighted_recall),
        float(weighted_f1),
        float(macro_f1),
        int(len(department_df)),
        int(len(model_df)),
        int(len(X_train)),
        int(len(X_test)),
        int(y.nunique()),
        int(minimum_records_per_department)
    ]
})

print("\nModel evaluation summary:")

display(evaluation_df)

labelled_data = model_df.copy()

labelled_data.insert(
    0,
    "labelled_record_id",
    range(
        1,
        len(labelled_data) + 1
    )
)

distribution_df = (
    model_df["department_label"]
    .value_counts()
    .rename_axis("department")
    .reset_index(name="record_count")
)

confusion_sql_df = (
    confusion_matrix_df
    .reset_index()
    .rename(
        columns={
            "index": "actual_department"
        }
    )
)

model_filename = (
    "04_department_prediction_model.pkl"
)

joblib.dump(
    department_prediction_model,
    model_filename
)

print("\nModel saved successfully:", model_filename)

prediction_results.to_csv(
    "04_department_prediction_results.csv",
    index=False
)

evaluation_df.to_csv(
    "04_department_model_evaluation.csv",
    index=False
)

report_df.to_csv(
    "04_department_classification_report.csv",
    index=False
)

labelled_data.to_csv(
    "04_department_labelled_data.csv",
    index=False
)

distribution_df.to_csv(
    "04_department_class_distribution.csv",
    index=False
)

confusion_sql_df.to_csv(
    "04_department_confusion_matrix.csv",
    index=False
)

print("\nCSV files saved successfully.")

prediction_results.to_sql(
    name="department_prediction_results",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

evaluation_df.to_sql(
    name="department_model_evaluation",
    con=database_engine,
    if_exists="replace",
    index=False
)

report_df.to_sql(
    name="department_classification_report",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=500
)

labelled_data.to_sql(
    name="department_labelled_data",
    con=database_engine,
    if_exists="replace",
    index=False,
    chunksize=1000
)

distribution_df.to_sql(
    name="department_class_distribution",
    con=database_engine,
    if_exists="replace",
    index=False
)

confusion_sql_df.to_sql(
    name="department_confusion_matrix",
    con=database_engine,
    if_exists="replace",
    index=False
)

print("\nAll Department Prediction tables created in MySQL.")

def predict_department(new_interaction):
    if not isinstance(new_interaction, str):
        raise TypeError(
            "The interaction must be entered as text."
        )

    cleaned_interaction = clean_interaction_text(
        new_interaction
    )

    if len(cleaned_interaction) < 3:
        raise ValueError(
            "Enter a meaningful government interaction."
        )

    predicted_department = (
        department_prediction_model.predict(
            [cleaned_interaction]
        )[0]
    )

    probabilities = (
        department_prediction_model.predict_proba(
            [cleaned_interaction]
        )[0]
    )

    probability_df = pd.DataFrame({
        "department": department_prediction_model.classes_,
        "probability_percentage": (
            probabilities * 100
        ).round(2)
    })

    probability_df = (
        probability_df
        .sort_values(
            by="probability_percentage",
            ascending=False
        )
        .reset_index(drop=True)
    )

    confidence = float(
        probability_df.iloc[0][
            "probability_percentage"
        ]
    )

    print("\nInteraction:", new_interaction)
    print(
        "Predicted department:",
        predicted_department
    )
    print(
        "Prediction confidence:",
        confidence,
        "%"
    )

    print("\nTop possible departments:")

    display(
        probability_df.head(5)
    )

    return {
        "interaction": new_interaction,
        "predicted_department": predicted_department,
        "confidence_percentage": confidence
    }

predict_department(
    "I need information about my railway ticket "
    "and train service complaint."
)

predict_department(
    "My hospital treatment benefit under the health "
    "scheme has not been approved."
)

predict_department(
    "There is a problem with my bank subsidy "
    "and benefit transfer."
)

predict_department(
    "I need help regarding a housing service "
    "and urban development issue."
)

predict_department(
    "My online government portal is not working "
    "and I cannot access the digital service."
)

verification_df = pd.DataFrame({
    "table_name": [
        "department_prediction_results",
        "department_model_evaluation",
        "department_classification_report",
        "department_labelled_data",
        "department_class_distribution",
        "department_confusion_matrix"
    ],
    "total_rows": [
        len(prediction_results),
        len(evaluation_df),
        len(report_df),
        len(labelled_data),
        len(distribution_df),
        len(confusion_sql_df)
    ]
})

print("\nMySQL tables created:")

display(verification_df)

sql_prediction_sample = pd.read_sql(
    """
    SELECT *
    FROM department_prediction_results
    LIMIT 10
    """,
    database_engine
)

print("\nSample predictions read from MySQL:")

display(sql_prediction_sample)

print("\n04_Department_Prediction completed successfully.")

print("\nSource table:", source_table)
print("Input column:", text_column)
print("Target column:", target_column)
print("Original records:", len(department_df))
print("Usable records:", len(model_df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Number of departments:", y.nunique())

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print(
    "Weighted F1-score:",
    round(weighted_f1 * 100, 2),
    "%"
)

print("\nSaved model:", model_filename)

print("\nMySQL output tables:")

print("1. department_prediction_results")
print("2. department_model_evaluation")
print("3. department_classification_report")
print("4. department_labelled_data")
print("5. department_class_distribution")
print("6. department_confusion_matrix")
```
